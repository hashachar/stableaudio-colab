```python
#@title 1 — Install
display(HTML("""
<div class="sa3-section">
  <h2>1 &mdash; Setup &amp; Installation</h2>
  <p>Run once per Colab session &middot; ~60–90 s without Flash Attention, ~2 min with &middot; restart not required</p>
</div>
"""))

import subprocess, sys

def _pip(*args):
    """Run pip install; print full output on failure so errors are visible."""
    result = subprocess.run(
        [sys.executable, "-m", "pip", "install", *args],
        capture_output=True,
        text=True,
    )
    if result.returncode != 0:
        print(result.stdout)
        print(result.stderr)
        raise RuntimeError(f"pip install failed for: {args}")

# stable-audio-3 is not on PyPI — install directly from the GitHub repo.
# --no-deps bypasses the torch==2.7.1 pin which conflicts with Colab's torch.
_pip("--quiet", "git+https://github.com/Stability-AI/stable-audio-3.git", "--no-deps")

# Non-torch runtime dependencies required by stable-audio-3.
_pip(
    "--quiet",
    "einops>=0.8.2",
    "soundfile>=0.13.1",
    "transformers>=5.8.0",
    "safetensors>=0.7.0",
    "huggingface-hub>=1.7.1",
    "numpy>=1.23",
)

# ipywidgets 8.x — Colab ships 7.x which lacks some Layout properties used below.
_pip("--quiet", "ipywidgets>=8.0")

# ─────────────────────────────────────────────────────────────────────────────
# Flash Attention 2 — required by the medium model for best performance.
# Only available on GPUs with compute capability ≥ 8.0 (A100, L4, H100).
#
# flash-attn ships pre-built wheels on GitHub only for specific torch versions.
# When Colab's torch is newer than any published wheel (currently the case —
# Colab runs torch 2.11 while the newest flash-attn wheel targets torch 2.9),
# NO pre-built wheel can match and the only option is a source build (~5–15 min).
#
# To avoid paying that build cost on every session, we cache the freshly-built
# wheel to Google Drive, keyed by a runtime fingerprint (python + cuda + torch).
# Later sessions reinstall from the cached wheel in ~30–60 s instead of rebuilding.
# If Colab upgrades torch/python the fingerprint changes and we rebuild once more.
# ─────────────────────────────────────────────────────────────────────────────

# ── Flash Attention options ──────────────────────────────────────────────────
# The standalone flash-attn package is OPTIONAL: stable-audio-3 imports it in a
# try/except and otherwise dispatches attention through torch's built-in
# F.scaled_dot_product_attention. On Ampere/Ada/Hopper GPUs (A100, L4, H100) in
# fp16/bf16 — which this notebook uses — PyTorch's SDPA already runs its own
# Flash-Attention-2 kernel, so single-prompt generation is at flash-attn speed
# WITHOUT the package. The package only adds variable-length packed attention,
# which the notebook's single-prompt path never triggers.
#
# Therefore the source build is OFF by default: on bleeding-edge torch (e.g.
# Colab's torch 2.11) no pre-built wheel exists and the source build both takes
# 5–15 min AND tends to fail (flash-attn predates the new torch C++ API). We
# still probe for a matching pre-built wheel and reuse any cached wheel, so if a
# compatible wheel ever exists it is used instantly.
BUILD_FLASH_ATTN_FROM_SOURCE = False  # True = attempt source build when no wheel (slow, may fail)
CACHE_FLASH_ATTN_WHEEL       = True   # Cache a built/downloaded wheel to Drive for reuse
FLASH_ATTN_WHEEL_DIR         = "/content/drive/MyDrive/stable-audio-cache/flash_attn_wheels"
# ─────────────────────────────────────────────────────────────────────────────

import torch as _torch
if _torch.cuda.is_available() and _torch.cuda.get_device_capability()[0] >= 8:
    import sys as _sys, os as _os, glob as _glob, shutil as _shutil

    _py = f"cp{_sys.version_info.major}{_sys.version_info.minor}"          # e.g. 'cp312'
    _cu = _torch.version.cuda.split('.')[0]                                # e.g. '12'
    _pt = '.'.join(_torch.__version__.split('+')[0].split('.')[:2])       # e.g. '2.11'
    _fingerprint = f"{_py}-cu{_cu}-torch{_pt}"

    def _flash_attn_importable():
        """True iff flash_attn imports cleanly against the current torch (checked in a
        fresh subprocess so a prior failed import in this kernel doesn't poison it)."""
        r = subprocess.run([sys.executable, "-c", "import flash_attn"],
                           capture_output=True, text=True)
        return r.returncode == 0

    def _pip_wheel(dest, *args):
        """Build a wheel into `dest`; print output on failure. Returns True on success."""
        r = subprocess.run([sys.executable, "-m", "pip", "wheel", *args, "-w", dest],
                           capture_output=True, text=True)
        if r.returncode != 0:
            print(r.stdout); print(r.stderr)
        return r.returncode == 0

    def _mount_drive():
        """Mount Google Drive if available; return True on success. Idempotent — a
        no-op if the Drive Cache cell already mounted it."""
        if _os.path.isdir("/content/drive/MyDrive"):
            return True
        try:
            from google.colab import drive
            drive.mount("/content/drive")
            return _os.path.isdir("/content/drive/MyDrive")
        except Exception:
            return False

    def _try_flash_attn_github():
        """Try every pre-built GitHub wheel matching this runtime (fast-fails ~1 s each)."""
        base     = "https://github.com/Dao-AILab/flash-attention/releases/download"
        versions = ["2.8.3", "2.8.2", "2.8.1", "2.8.0",
                    "2.7.4.post1", "2.7.3", "2.7.2", "2.7.1", "2.6.3", "2.6.2"]
        for ver in versions:
            for cxx in ("FALSE", "TRUE"):
                whl = (f"flash_attn-{ver}+cu{_cu}torch{_pt}"
                       f"cxx11abi{cxx}-{_py}-{_py}-linux_x86_64.whl")
                url = f"{base}/v{ver}/{whl}"
                try:
                    _pip("--quiet", url)
                    return True
                except RuntimeError:
                    continue
        return False

    def _install_cached_wheel():
        """Install a previously-built wheel from Drive if one matches this fingerprint."""
        if not (CACHE_FLASH_ATTN_WHEEL and _mount_drive()):
            return False
        cache_dir = _os.path.join(FLASH_ATTN_WHEEL_DIR, _fingerprint)
        whls = sorted(_glob.glob(_os.path.join(cache_dir, "flash_attn-*.whl")))
        if not whls:
            return False
        try:
            _pip("--quiet", whls[-1])
        except RuntimeError:
            return False
        if _flash_attn_importable():
            return True
        # Stale / ABI-incompatible wheel — drop it so the build step re-caches a good one.
        try:
            _os.remove(whls[-1])
        except OSError:
            pass
        return False

    def _build_and_cache_wheel():
        """Build flash-attn from source once, install it, and cache the wheel to Drive."""
        # ninja makes the CUDA compile dramatically faster (minutes vs. far longer);
        # MAX_JOBS caps parallel compile jobs so the build doesn't exhaust Colab RAM.
        try:
            _pip("--quiet", "ninja", "packaging", "psutil")
        except RuntimeError:
            pass
        _os.environ.setdefault("MAX_JOBS", "4")
        build_dir = "/content/_flash_attn_build"
        _os.makedirs(build_dir, exist_ok=True)
        if not _pip_wheel(build_dir, "flash-attn", "--no-build-isolation", "--no-deps"):
            return False
        whls = sorted(_glob.glob(_os.path.join(build_dir, "flash_attn-*.whl")))
        if not whls:
            return False
        try:
            _pip("--quiet", whls[-1])
        except RuntimeError:
            return False
        # Cache the built wheel for future sessions.
        if CACHE_FLASH_ATTN_WHEEL and _mount_drive():
            cache_dir = _os.path.join(FLASH_ATTN_WHEEL_DIR, _fingerprint)
            try:
                _os.makedirs(cache_dir, exist_ok=True)
                _shutil.copy2(whls[-1], cache_dir)
            except Exception:
                pass
        return _flash_attn_importable()

    # Resolution order: already installed → pre-built GitHub wheel → Drive cache → build.
    if _flash_attn_importable():
        display(HTML('<div class="sa3-info" style="margin:4px 0;">⚡ Flash Attention 2 already installed.</div>'))
    elif _try_flash_attn_github():
        display(HTML('<div class="sa3-info" style="margin:4px 0;">⚡ Flash Attention 2 installed (pre-built wheel).</div>'))
    elif _install_cached_wheel():
        display(HTML('<div class="sa3-info" style="margin:4px 0;">⚡ Flash Attention 2 installed from Drive cache &mdash; no rebuild needed.</div>'))
    elif not BUILD_FLASH_ATTN_FROM_SOURCE:
        # No pre-built/cached wheel and source build disabled → use torch SDPA,
        # which already runs a Flash-Attention-2 kernel on this GPU in fp16/bf16.
        display(HTML(
            '<div style="font-size:12px;color:#6B7280;margin:2px 0;">'
            'No pre-built Flash Attention 2 wheel for this runtime '
            f'(<code>{_fingerprint}</code>) &mdash; using PyTorch&rsquo;s built-in '
            'Flash-Attention kernel via <code>scaled_dot_product_attention</code>, '
            'which delivers the same speed for single-prompt generation on this GPU. '
            'Set <code>BUILD_FLASH_ATTN_FROM_SOURCE = True</code> to attempt a '
            '(slow, ~5–15 min) source build of the standalone package instead.'
            '</div>'
        ))
    else:
        display(HTML(
            '<div class="sa3-warn" style="margin:4px 0;">'
            '⏳ No pre-built or cached Flash Attention 2 wheel for this runtime '
            f'(<code>{_fingerprint}</code>). Building from source once &mdash; this takes '
            '<strong>5–15 min</strong> and may fail on bleeding-edge torch. On success '
            'the wheel is cached to Google Drive so future sessions install it in ~30–60 s.'
            '</div>'
        ))
        if _build_and_cache_wheel():
            display(HTML('<div class="sa3-info" style="margin:4px 0;">⚡ Flash Attention 2 built, installed, and cached to Drive for next time.</div>'))
        else:
            display(HTML(
                '<div style="font-size:12px;color:#6B7280;margin:2px 0;">'
                'Flash Attention 2 could not be built &mdash; skipped. The medium model '
                'still runs via PyTorch&rsquo;s built-in Flash-Attention SDPA kernel.'
                '</div>'
            ))
else:
    display(HTML(
        '<div style="font-size:12px;color:#6B7280;margin:2px 0;">'
        'Flash Attention 2 skipped — requires compute capability ≥ 8.0 (A100 / L4 / H100).'
        '</div>'
    ))

# NOTE: stable_audio_3 / huggingface_hub are NOT imported here on purpose.
# huggingface_hub reads HF_HOME once at import time; importing it before the
# Drive cache cell would lock the cache path to the ephemeral /root/.cache
# and ignore the Google Drive path set in the next cell.
import torch, torchaudio

display(HTML(f"""
<div class="sa3-info" style="margin-top:12px;">
  <strong>Installation complete.</strong><br>
  torch {torch.__version__} &nbsp;|&nbsp;
  torchaudio {torchaudio.__version__} &nbsp;|&nbsp;
  CUDA available: <strong>{torch.cuda.is_available()}</strong>
  {'&nbsp;|&nbsp; GPU: ' + torch.cuda.get_device_name(0) if torch.cuda.is_available() else ''}
</div>
"""))
```

```python
#@title 4 — Utilities
display(HTML("""
<div class="sa3-section">
  <h2>4 &mdash; Utilities <span style="font-weight:400;opacity:.75;font-size:14px;">(run once)</span></h2>
  <p>Model registry, in-memory cache, and the shared generation function used throughout the notebook.</p>
</div>
"""))

import os, time, tempfile, warnings
import torch, torchaudio
from stable_audio_3 import StableAudioModel

# ── Cold-start controls ──────────────────────────────────────────────────────
# ENABLE_TORCH_COMPILE: torch.compile JIT-compiles the DiT on the FIRST generate()
# call. That compile is the multi-minute stall on the first run of a fresh runtime
# (and it recompiles whenever the clip duration — hence latent length — changes).
# For an 8-step sampler whose per-step cost is already Flash-Attention via SDPA, the
# payoff is small and rarely recouped in an interactive demo, so it is OFF by default.
# Set True for many same-duration generations; we then use dynamic=True (one compile,
# no per-duration recompiles) and a persistent on-Drive inductor cache.
ENABLE_TORCH_COMPILE = False
# WARMUP_ON_LOAD: run one throwaway 2 s generation right after the model loads, so the
# unavoidable first-call costs (CUDA context, cuDNN/cuBLAS, SDPA kernel pick, T5 first
# matmul, VAE first decode) are paid during the expected 'Loading…' wait instead of on
# your first real prompt.
WARMUP_ON_LOAD = True

# stable_audio_3 calls hf_hub_download via a module-level `from huggingface_hub
# import hf_hub_download` — a local name reference that bypasses huggingface_hub's
# own token auto-detection on Colab. We replace that reference with a thin wrapper
# that injects the token from the environment whenever it isn't already provided.
import stable_audio_3.model_configs as _sa3_cfg
# Always source the real hf_hub_download from the huggingface_hub module,
# never from _sa3_cfg (which may already be our wrapper). This makes the
# patch idempotent: re-running the cell just replaces the wrapper with an
# identical one that still calls the real function — no recursion possible.
import huggingface_hub as _hf_hub
_real_hf_hub_download = _hf_hub.hf_hub_download
def _sa3_hf_hub_download_authed(*args, **kwargs):
    if not kwargs.get("token"):
        kwargs["token"] = os.environ.get("HF_TOKEN") or None
    return _real_hf_hub_download(*args, **kwargs)
_sa3_cfg.hf_hub_download = _sa3_hf_hub_download_authed

# Patch load_diffusion_cond for two reasons:
#  1. Load state dict to CPU so copy_state_dict is CPU→CPU (fast), then a single
#     model.to(device) transfers everything to GPU in one batch. The previous approach
#     (load to GPU, copy_state_dict into CPU model) caused one GPU→CPU sync per tensor
#     — hundreds of small transfers — producing the silent 2-minute wait.
#  2. Force fp16 on CUDA: A100/L4/H100 Tensor Cores are 2–3× faster in fp16 than fp32.
import stable_audio_3.loading_utils as _sa3_lu
import stable_audio_3.factory as _sa3_factory
import gc as _gc
_orig_load_diffusion_cond = _sa3_lu.load_diffusion_cond
def _load_diffusion_cond_fast(model_config, ckpt_path, device='cuda', model_half=False):
    from safetensors.torch import load_file as _lf
    _dev = device if isinstance(device, str) else str(device)
    if not model_half and _dev.startswith('cuda'):
        model_half = True   # fp16 on A100/L4/H100: 2–3× faster inference
    model = _sa3_factory.create_diffusion_cond_from_config(model_config)
    state_dict = _lf(ckpt_path)   # CPU; copy_state_dict is then CPU→CPU (fast)
    _sa3_lu.copy_state_dict(model, state_dict)
    del state_dict
    _gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
    model.to(device).eval().requires_grad_(False)
    if model_half:
        model.to(torch.float16)
    return model
_sa3_lu.load_diffusion_cond = _load_diffusion_cond_fast

SAMPLE_RATE = 44_100

MODEL_CONFIGS = {
    "small-music": {
        "hf_name":       "small-music",
        "max_duration":  120.0,
        "default_steps": 8,
        "default_cfg":   1.0,
        "needs_gpu":     False,
        "label":         "small-music  —  433 M params, CPU-ok, max 120 s",
    },
    "medium": {
        "hf_name":       "medium",
        "max_duration":  285.0,
        "default_steps": 8,
        "default_cfg":   1.0,
        "needs_gpu":     True,
        "label":         "medium  —  1.4 B params, GPU required, max 285 s",
    },
}

_model_cache: dict = {}

# Shared session audio registry — maps user-given title → wav path in /tmp.
# Populated by the "Save to session" button in section 5.
# Read by the source dropdown in section 6.
_session_audio: dict = {}

# Conditioning cache keyed on (id(model), prompt_str, round(duration, 2)).
# RF-Inversion reuses T5-encoded conditioning across repeated calls with the same prompt.
_cond_cache: dict = {}

# Session-list change notification. Each tool cell registers a listener
# (tagged, so re-running a cell replaces its listener instead of stacking a
# duplicate) to refresh its source dropdown the moment ANY tool saves a clip.
_session_audio_listeners: dict = {}

def _add_session_listener(tag, callback):
    _session_audio_listeners[tag] = callback

def _on_session_audio_updated():
    for _cb in list(_session_audio_listeners.values()):
        try:
            _cb()
        except Exception:
            pass

def _register_session_clip(base_title, src_path):
    """Copy a result wav to a unique /tmp path and add it to the session list.

    Tools write results to fixed paths (e.g. /tmp/_sa3_a2a_result.wav) that the
    next run overwrites, so the registry keeps its own copy. Returns the title
    actually used (suffixed if the base title is taken).
    """
    import shutil
    title = base_title
    i = 2
    while title in _session_audio:
        title = f"{base_title} ({i})"
        i += 1
    dst = tempfile.mktemp(suffix=".wav", dir="/tmp")
    shutil.copyfile(src_path, dst)
    _session_audio[title] = dst
    _on_session_audio_updated()
    return title


def _is_model_on_disk(model_name: str) -> bool:
    """Return True if the model checkpoint is already in the local HF cache."""
    try:
        from huggingface_hub import try_to_load_from_cache
        repo_id = _sa3_cfg.all_models[MODEL_CONFIGS[model_name]["hf_name"]].repo_id
        result = try_to_load_from_cache(repo_id=repo_id, filename="model.safetensors")
        return isinstance(result, str)
    except Exception:
        return False


def get_model(model_name: str) -> StableAudioModel:
    if model_name not in _model_cache:
        # Free any other cached model before loading to reclaim GPU VRAM
        for _other in list(_model_cache):
            if _other != model_name:
                del _model_cache[_other]
        _gc.collect()
        if torch.cuda.is_available():
            torch.cuda.empty_cache()
        cfg = MODEL_CONFIGS[model_name]
        if cfg["needs_gpu"] and not torch.cuda.is_available():
            raise RuntimeError(
                "The 'medium' model requires a CUDA GPU. "
                "Go to Runtime → Change runtime type → GPU."
            )
        if cfg["needs_gpu"] and torch.cuda.is_available():
            cc = torch.cuda.get_device_capability()
            if cc[0] < 8:
                display(HTML(
                    f'<div class="sa3-warn" style="margin:6px 0;">'
                    f'&#9888;&#65039;&nbsp; Flash Attention 2 is not available on this GPU '
                    f'(compute capability {cc[0]}.{cc[1]}, needs &ge;8.0). '
                    f'<strong>medium</strong> will still work — generation will just be '
                    f'slower than on an A100 or L4.'
                    f'</div>'
                ))
        if not os.environ.get("HF_TOKEN"):
            raise RuntimeError(
                "HF_TOKEN not found in environment. "
                "Run section 3 (Hugging Face Authentication) and make sure "
                "the HF_TOKEN Colab secret exists with Notebook access enabled."
            )
        device = "cuda" if torch.cuda.is_available() else "cpu"
        print(f"Loading '{model_name}' on {device}…")
        print("  Phase 1/2: diffusion model weights (progress bar below)")
        print("  Phase 2/2: VAE + text encoder — no progress bar, expect 60–90 s")

        import threading as _threading
        _stop_ticker = _threading.Event()
        def _ticker():
            t0 = time.time()
            while not _stop_ticker.wait(30):
                print(f"  Still loading… {time.time() - t0:.0f}s elapsed")
        _threading.Thread(target=_ticker, daemon=True).start()
        try:
            _model_cache[model_name] = StableAudioModel.from_pretrained(
                cfg["hf_name"], device=device
            )
        except Exception as exc:
            exc_str = str(exc)
            if "401" in exc_str or "Unauthorized" in exc_str or "GatedRepo" in type(exc).__name__:
                raise RuntimeError(
                    "Authentication failed (401 Unauthorized). "
                    "Your HF_TOKEN may be invalid or the model license has not been accepted. "
                    "Visit https://huggingface.co/stabilityai/stable-audio-3-small-music "
                    "to accept the license, then re-run section 3."
                ) from exc
            raise
        finally:
            _stop_ticker.set()
        if ENABLE_TORCH_COMPILE and torch.cuda.is_available() and hasattr(_model_cache[model_name], 'model'):
            import torch._dynamo as _dynamo
            _dynamo.config.suppress_errors = True
            # Persist compiled kernels on Drive so the JIT cost is paid at most once
            # across sessions, not once per runtime.
            _ind_cache = '/content/drive/MyDrive/stable-audio-cache/torchinductor'
            try:
                os.makedirs(_ind_cache, exist_ok=True)
                os.environ.setdefault('TORCHINDUCTOR_CACHE_DIR', _ind_cache)
            except Exception:
                pass
            # dynamic=True → a single compile that handles every duration, instead of a
            # fresh recompile each time the clip length changes.
            _model_cache[model_name].model = torch.compile(
                _model_cache[model_name].model, mode="default", dynamic=True
            )
            print("  DiT compiled (torch.compile, dynamic) — first generate() triggers a one-time JIT")

        # Warmup: pay the one-time first-call CUDA/cuDNN/SDPA/T5/VAE costs now, behind the
        # 'Loading…' wait, rather than on the user's first real prompt. Also triggers the
        # torch.compile JIT here when compile is enabled.
        if WARMUP_ON_LOAD and torch.cuda.is_available():
            print("  Warming up kernels (one-time)…")
            try:
                _t0 = time.time()
                with torch.inference_mode():
                    _model_cache[model_name].generate(
                        prompt="warmup", duration=2.0, steps=4,
                        cfg_scale=1.0, seed=0, chunked_decode=True,
                    )
                torch.cuda.synchronize()
                print(f"  Warmup done in {time.time() - _t0:.0f}s — first real generation runs at full speed.")
            except Exception as _werr:
                print(f"  (warmup skipped: {_werr})")
        print(f"'{model_name}' ready.")
    return _model_cache[model_name]


print("Utilities loaded.")
```

```python
#@title generate_audio()
def generate_audio(
    prompt: str,
    duration_seconds: float,
    model_name: str = "small-music",
    seed: int = -1,
    steps: int | None = None,
    cfg_scale: float | None = None,
    negative_prompt: str | None = None,
) -> tuple[str, float]:
    """
    Generate audio from a text prompt.

    Returns
    -------
    (wav_path, elapsed_seconds)
        wav_path        Absolute path to the saved WAV file in /tmp.
        elapsed_seconds Wall-clock generation time.

    This function is the single text-to-music entry point used by the
    interactive demo (section 5).
    """
    cfg = MODEL_CONFIGS[model_name]
    duration_seconds = min(duration_seconds, cfg["max_duration"])
    steps     = steps     if steps     is not None else cfg["default_steps"]
    cfg_scale = cfg_scale if cfg_scale is not None else cfg["default_cfg"]

    model = get_model(model_name)

    t0 = time.perf_counter()
    with torch.inference_mode():
        audio_tensor = model.generate(
            prompt=prompt,
            negative_prompt=negative_prompt,
            duration=duration_seconds,
            steps=steps,
            cfg_scale=cfg_scale,
            seed=seed,
            chunked_decode=True,
        )
    elapsed = time.perf_counter() - t0

    audio_tensor = audio_tensor[0]  # (channels, samples) — drop batch dim
    tmp_path = tempfile.mktemp(suffix=".wav", dir="/tmp")
    torchaudio.save(tmp_path, audio_tensor.cpu(), SAMPLE_RATE)
    return tmp_path, elapsed


print("generate_audio() ready.")
```

```python
#@title 5 — Generate
display(HTML("""
<div class="sa3-section">
  <h2>5 &mdash; Generate Music</h2>
  <p>Text prompt + duration only &mdash; no audio reference input.</p>
</div>
"""))

# ── Widgets ───────────────────────────────────────────────────────────────────
prompt_textarea = widgets.Textarea(
    value=("Late-night jazz trio: upright bass walking line, brushed snare, "
           "and a melancholy piano melody with blue notes and subtle chord extensions"),
    placeholder="Describe the audio you want to generate\u2026",
    layout=Layout(width="100%", height="72px"),
    style={"description_width": "0px"},
)

model_dropdown = widgets.Dropdown(
    options=[(MODEL_CONFIGS[k]["label"], k) for k in MODEL_CONFIGS],
    value="small-music",
    description="Model:",
    style={"description_width": "52px"},
    layout=Layout(width="500px"),
)

duration_slider = widgets.FloatSlider(
    value=30.0,
    min=5.0,
    max=MODEL_CONFIGS["small-music"]["max_duration"],
    step=5.0,
    description="Duration:",
    style={"description_width": "52px"},
    layout=Layout(width="440px"),
    readout_format=".0f",
)

gpu_warning = widgets.HTML(
    value='<div class="sa3-warn">'
          '&#9888;&#65039;&nbsp; <b>medium</b> requires a GPU with compute capability ≥8.0 '
          '(A100 or L4). No compatible GPU detected. '
          'Go to <b>Runtime &rarr; Change runtime type &rarr; GPU</b>.'
          '</div>',
    layout=Layout(display="none", width="100%"),
)

# ── Advanced options (collapsed by default — defaults match the model's tuning) ──
adv_neg_prompt = widgets.Textarea(
    value="", placeholder="Negative prompt — qualities to steer away from (e.g. 'distortion, vocals')…",
    layout=Layout(width="490px", height="52px"),
)
adv_steps_slider = widgets.IntSlider(
    value=8, min=4, max=50, step=1,
    description="Steps:", style={"description_width": "52px"},
    layout=Layout(width="440px"), continuous_update=False,
)
adv_cfg_slider = widgets.FloatSlider(
    value=1.0, min=0.0, max=15.0, step=0.5,
    description="CFG:", style={"description_width": "52px"},
    layout=Layout(width="440px"), readout_format=".1f", continuous_update=False,
)
adv_seed = widgets.BoundedIntText(
    value=-1, min=-1, max=2**31 - 1,
    description="Seed:", style={"description_width": "52px"},
    layout=Layout(width="200px"),
)
adv_takes_slider = widgets.IntSlider(
    value=1, min=1, max=4, step=1,
    description="Takes:", style={"description_width": "52px"},
    layout=Layout(width="300px"), continuous_update=False,
)
advanced_box = widgets.Accordion(
    children=[widgets.VBox([
        widgets.HTML(
            '<div style="font-size:12px;color:#6B7280;margin:2px 0 3px;">'
            '<b>Negative prompt</b> '
            '<span style="color:#9CA3AF;">— what the output should avoid</span></div>'
        ),
        adv_neg_prompt,
        widgets.HTML(
            '<div style="font-size:12px;color:#6B7280;margin:10px 0 1px;">'
            '<b>Steps</b> '
            '<span style="color:#9CA3AF;">— denoising steps. The model is tuned for 8; '
            'more refines detail with diminishing returns and roughly linear cost.</span></div>'
        ),
        adv_steps_slider,
        widgets.HTML(
            '<div style="font-size:12px;color:#6B7280;margin:10px 0 1px;">'
            '<b>CFG scale</b> '
            '<span style="color:#9CA3AF;">— prompt adherence. 1.0 = model default; '
            '2&ndash;4 = stronger steering; 7+ = literal but can sound forced.</span></div>'
        ),
        adv_cfg_slider,
        widgets.HTML(
            '<div style="font-size:12px;color:#6B7280;margin:10px 0 1px;">'
            '<b>Seed</b> '
            '<span style="color:#9CA3AF;">— &minus;1 = random each run; any other value '
            'reproduces the same output for the same settings.</span></div>'
        ),
        adv_seed,
        widgets.HTML(
            '<div style="font-size:12px;color:#6B7280;margin:10px 0 1px;">'
            '<b>Takes</b> '
            '<span style="color:#9CA3AF;">&mdash; generate several takes of the same '
            'prompt in one click, each with its own seed.</span></div>'
        ),
        adv_takes_slider,
    ], layout=Layout(padding="6px 2px"))],
    selected_index=None,
    layout=Layout(width="500px", margin="4px 0"),
)
advanced_box.set_title(0, "Advanced options")

generate_btn = widgets.Button(
    description="  Generate Audio",
    icon="music",
    button_style="",
    layout=Layout(width="190px", height="42px"),
)
generate_btn.style.button_color = "#7C3AED"
generate_btn.style.font_weight = "700"

output_area = widgets.Output(layout=Layout(min_height="60px", margin="6px 0 0"))

SPINNER_HTML = """
<div style="display:flex;align-items:center;gap:14px;padding:14px 0;
            font-family:-apple-system,BlinkMacSystemFont,'Segoe UI',sans-serif;">
  <div style="width:26px;height:26px;border-radius:50%;
              border:3px solid #EDE9FE;border-top-color:#7C3AED;
              animation:sa3-spin .7s linear infinite,sa3-pulse 1.4s ease-in-out infinite;
              flex-shrink:0;"></div>
  <div>
    <div style="font-weight:600;color:#5B21B6;font-size:14px;">{status}</div>
    <div style="font-size:12px;color:#9CA3AF;margin-top:2px;">{substatus}</div>
  </div>
</div>
"""

# ── Callbacks ─────────────────────────────────────────────────────────────────
def on_model_change(change):
    new_model = change["new"]
    cfg = MODEL_CONFIGS[new_model]
    duration_slider.max = cfg["max_duration"]   # max before value to avoid silent clamp
    if duration_slider.value > cfg["max_duration"]:
        duration_slider.value = cfg["max_duration"]
    needs_warn = cfg["needs_gpu"] and not torch.cuda.is_available()
    gpu_warning.layout.display = "" if needs_warn else "none"

model_dropdown.observe(on_model_change, names="value")


def _unique_session_title(base):
    """Return base, or base (2), base (3)… if already taken."""
    if base not in _session_audio:
        return base
    i = 2
    while f"{base} ({i})" in _session_audio:
        i += 1
    return f"{base} ({i})"


def on_generate_clicked(_btn):
    prompt     = prompt_textarea.value.strip()
    duration   = duration_slider.value
    model_name = model_dropdown.value

    if not prompt:
        with output_area:
            clear_output(wait=True)
            display(HTML('<div class="sa3-error">Please enter a prompt before generating.</div>'))
        return

    generate_btn.disabled = True
    generate_btn.description = "  Generating…"

    if model_name in _model_cache:
        status    = "Running diffusion…"
        substatus = f"&#x26A1;&nbsp; <strong>{model_name}</strong> already loaded in memory."
    elif _is_model_on_disk(model_name):
        status    = f"Loading <strong>{model_name}</strong> from Drive cache…"
        substatus = "&#x1F4BE;&nbsp; Weights found on Drive — no download needed."
    else:
        status    = f"Downloading <strong>{model_name}</strong> weights…"
        substatus = "&#x2B07;&nbsp; First run — may take 1&ndash;3&thinsp;min depending on connection."

    with output_area:
        clear_output(wait=True)
        display(HTML(SPINNER_HTML.format(status=status, substatus=substatus)))

    try:
        import random as _random
        n_takes   = adv_takes_slider.value
        base_seed = adv_seed.value
        # Explicit seeds even when the user asked for random: every result panel
        # can then show the seed that made it, so any take is reproducible.
        seeds = ([_random.randint(0, 2**31 - 1) for _ in range(n_takes)]
                 if base_seed == -1 else [base_seed + i for i in range(n_takes)])
        short = prompt[:90] + ("…" if len(prompt) > 90 else "")

        takes = []  # (path, elapsed, seed)
        for _i, _s in enumerate(seeds, start=1):
            if n_takes > 1:
                with output_area:
                    clear_output(wait=True)
                    display(HTML(SPINNER_HTML.format(
                        status=f"Generating take {_i}/{n_takes}…",
                        substatus=f"seed {_s}",
                    )))
            _path, _elapsed = generate_audio(
                prompt, duration, model_name,
                seed=_s,
                steps=adv_steps_slider.value,
                cfg_scale=adv_cfg_slider.value,
                negative_prompt=adv_neg_prompt.value.strip() or None,
            )
            takes.append((_path, _elapsed, _s))
        path, elapsed = takes[0][0], takes[0][1]

        # Auto-register immediately so every tool's source dropdown updates at once
        _base = (prompt[:40].strip()) or "Generated audio"
        for _i, (_p, _, _s) in enumerate(takes, start=1):
            _t = _unique_session_title(_base if n_takes == 1 else f"{_base} · take {_i}")
            if _i == 1:
                _auto_title = _t
            _session_audio[_t] = _p
        _on_session_audio_updated()

        # ── Optional rename UI ─────────────────────────────────────────────────
        save_title_input = widgets.Text(
            value=_auto_title,
            placeholder="Rename this clip…",
            layout=Layout(width="320px"),
        )
        save_btn_inner = widgets.Button(
            description="Rename",
            icon="pencil",
            layout=Layout(width="110px", height="34px"),
        )
        save_btn_inner.style.button_color = "#EDE9FE"
        save_status_lbl = widgets.HTML()

        def on_save(b):
            new_title = save_title_input.value.strip()
            if not new_title:
                save_status_lbl.value = (
                    '<span style="color:#DC2626;font-size:13px;">Title cannot be empty.</span>'
                )
                return
            if new_title != _auto_title:
                _session_audio.pop(_auto_title, None)
                _session_audio[new_title] = path
                _on_session_audio_updated()
            save_status_lbl.value = (
                f'<span style="color:#059669;font-size:13px;">'
                f'&#x2713;&nbsp; &ldquo;{new_title}&rdquo;</span>'
            )
            save_btn_inner.disabled = True

        save_btn_inner.on_click(on_save)

        save_row = widgets.VBox([
            widgets.HTML(
                '<div style="font-size:12px;color:#6B7280;margin:12px 0 3px;">'
                f'Saved as &ldquo;<strong>{_auto_title}</strong>&rdquo; — rename if you like:</div>'
            ),
            widgets.HBox(
                [save_title_input, save_btn_inner, save_status_lbl],
                layout=Layout(align_items="center", gap="8px"),
            ),
        ])

        with output_area:
            clear_output(wait=True)
            display(HTML(f"""
            <div class="sa3-result">
              <div style="display:flex;align-items:center;gap:8px;margin-bottom:8px;flex-wrap:wrap;">
                <span class="sa3-badge">{model_name}</span>
                <span class="sa3-stat">{duration:.0f}&thinsp;s</span>
                <span class="sa3-stat">generated in {sum(t[1] for t in takes):.1f}&thinsp;s</span>
              </div>
              <div style="font-size:13px;color:#374151;font-style:italic;line-height:1.5;">
                &ldquo;{short}&rdquo;
              </div>
            </div>
            """))
            for _i, (_p, _el, _s) in enumerate(takes, start=1):
                if n_takes > 1:
                    display(HTML(
                        f'<div style="font-size:12px;color:#6B7280;margin:10px 0 2px;">'
                        f'<b>Take {_i}/{n_takes}</b>'
                        f'<span class="sa3-stat" style="margin-left:8px;">seed {_s}</span>'
                        f'<span class="sa3-stat" style="margin-left:6px;">{_el:.1f}&thinsp;s</span></div>'
                    ))
                else:
                    display(HTML(
                        f'<div style="font-size:12px;color:#9CA3AF;margin:6px 0 2px;">'
                        f'seed {_s} &mdash; enter it under Advanced options to reproduce this exact result.</div>'
                    ))
                display(Audio(_p, autoplay=False))
                _dl = widgets.Button(description="  Download", icon="download",
                                     layout=Layout(width="150px", height="34px"))
                _dl.style.button_color = "#EDE9FE"
                def _dl_click(b, _pp=_p):
                    from google.colab import files as _cf
                    _cf.download(_pp)
                _dl.on_click(_dl_click)
                display(_dl)
            if n_takes == 1:
                display(save_row)
            else:
                display(HTML(
                    '<div style="font-size:12px;color:#9CA3AF;margin:10px 0 0;">'
                    'All takes saved to session clips &mdash; usable as sources in every tool.</div>'
                ))

    except RuntimeError as exc:
        with output_area:
            clear_output(wait=True)
            display(HTML(f'<div class="sa3-error"><strong>Error:</strong> {exc}</div>'))
    finally:
        generate_btn.disabled = False
        generate_btn.description = "  Generate Audio"


generate_btn.on_click(on_generate_clicked)

# ── Layout ────────────────────────────────────────────────────────────────────
ui = widgets.VBox(
    [
        widgets.HTML("<div style='font-size:12px;color:#6B7280;margin:0 0 4px;"
                     "font-family:sans-serif;'>Prompt:</div>"),
        prompt_textarea,
        widgets.HBox([model_dropdown]),
        gpu_warning,
        widgets.HBox(
            [duration_slider, widgets.Label("seconds", layout=Layout(margin="0 0 0 8px"))],
            layout=Layout(align_items="center"),
        ),
        advanced_box,
        generate_btn,
        output_area,
    ],
    layout=Layout(width="100%", max_width="700px"),
)
display(ui)
```

```python
#@title 6 — Inpainting
display(HTML("""
<div class="sa3-section">
  <h2>6 &mdash; Inpainting / Audio Editing</h2>
  <p>
    Replace a time segment of existing audio with new content described by a prompt.
    Works with both <strong>small-music</strong> (CPU-ok) and <strong>medium</strong> (GPU required).
  </p>
</div>
"""))

# ── Imports ───────────────────────────────────────────────────────────────────
import matplotlib.pyplot as plt
import numpy as np
from io import BytesIO
from IPython.display import Image
from google.colab import files as colab_files

# All other names (widgets, Layout, Audio, clear_output, HTML, display,
# torchaudio, SAMPLE_RATE, time, tempfile, get_model, SPINNER_HTML,
# _session_audio, _on_session_audio_updated) come from earlier cells.


# ── Waveform renderer ─────────────────────────────────────────────────────────
def _render_waveform(audio_tensor, mask_start, mask_end, view_start=None, view_end=None):
    """Render waveform PNG. view_start/view_end zoom the x-axis."""
    mono = audio_tensor.mean(dim=0).numpy()
    duration = len(mono) / SAMPLE_RATE
    step = max(1, len(mono) // 8000)
    mono_ds = mono[::step]
    times = np.linspace(0, duration, len(mono_ds))
    mask_start = max(0.0, float(mask_start))
    mask_end   = min(duration, float(mask_end))
    v0 = max(0.0, float(view_start)) if view_start is not None else 0.0
    v1 = min(duration, float(view_end)) if view_end is not None else duration
    if v0 >= v1:
        v0, v1 = 0.0, duration

    fig, ax = plt.subplots(figsize=(9, 2.4), facecolor='#F5F3FF')
    ax.set_facecolor('#F5F3FF')
    ax.plot(times, mono_ds, color='#7C3AED', linewidth=0.4, alpha=0.9, rasterized=True)
    ax.fill_between(times, mono_ds, alpha=0.18, color='#7C3AED')
    ax.axvspan(mask_start, mask_end, alpha=0.45, color='#FCD34D', zorder=2)
    ax.axvline(mask_start, color='#D97706', linewidth=2.0, zorder=3)
    ax.axvline(mask_end,   color='#D97706', linewidth=2.0, zorder=3)
    mid = (mask_start + mask_end) / 2
    if v0 <= mid <= v1:
        y_top = float(np.abs(mono_ds).max()) * 0.88 if len(mono_ds) > 0 else 0.8
        ax.text(mid, y_top,
                f'{mask_start:.2f}s\u2013{mask_end:.2f}s  ({mask_end - mask_start:.2f}s)',
                ha='center', va='top', fontsize=8, color='#92400E',
                bbox=dict(boxstyle='round,pad=0.3', facecolor='#FEF3C7',
                          edgecolor='#FCD34D', alpha=0.95), zorder=4)
    ax.set_xlim(v0, v1)
    ax.set_xlabel('Time (s)', fontsize=9, color='#374151')
    ax.tick_params(colors='#6B7280', labelsize=8)
    for sp in ax.spines.values():
        sp.set_edgecolor('#DDD6FE')
    plt.tight_layout(pad=0.4)
    buf = BytesIO()
    fig.savefig(buf, format='png', dpi=110, bbox_inches='tight', facecolor='#F5F3FF')
    plt.close(fig)
    buf.seek(0)
    return buf.read()


# ── State ─────────────────────────────────────────────────────────────────────
_inpaint_audio = None   # torch.Tensor (channels, samples) at SAMPLE_RATE


# ── Widgets ───────────────────────────────────────────────────────────────────
_UPLOAD_OPT = '\u2014 Upload new file \u2014'

inpaint_model_dropdown = widgets.Dropdown(
    options=[(MODEL_CONFIGS[k]["label"], k) for k in MODEL_CONFIGS],
    value="small-music",
    description="Model:",
    style={"description_width": "52px"},
    layout=Layout(width="500px"),
)

inpaint_gpu_warning = widgets.HTML(
    value='<div class="sa3-warn">'
          '&#9888;&#65039;&nbsp; <b>medium</b> requires a GPU with compute capability \u22658.0 '
          '(A100 or L4). No compatible GPU detected. '
          'Go to <b>Runtime &rarr; Change runtime type &rarr; GPU</b>.'
          '</div>',
    layout=Layout(display="none", width="100%"),
)

def _limit_note_html(model_name):
    cap = MODEL_CONFIGS[model_name]['max_duration']
    return (
        '<div style="font-size:12px;color:#4B5563;background:#F3F4F6;'
        'border-left:3px solid #9CA3AF;padding:6px 10px;border-radius:3px;margin:2px 0;">'
        f'&#8505;&#65039;&nbsp; <b>{model_name}</b> processes only the first '
        f'<b>{cap:.0f}&thinsp;s</b> of the source. '
        f'Place the mask within 0&thinsp;s&ndash;{cap:.0f}&thinsp;s.'
        '</div>'
    )

inpaint_limit_note = widgets.HTML(
    value=_limit_note_html('small-music'),
    layout=Layout(width='100%'),
)

source_dropdown = widgets.Dropdown(
    options=[_UPLOAD_OPT],
    description='Source:',
    style={'description_width': '60px'},
    layout=Layout(width='440px'),
)

upload_btn = widgets.Button(
    description='Upload Audio', icon='upload',
    layout=Layout(width='155px', height='36px'),
)
upload_btn.style.button_color = '#EDE9FE'

upload_status = widgets.HTML(
    value='<div class="sa3-info" style="margin:4px 0;">No file loaded yet.</div>'
)
upload_row = widgets.HBox(
    [upload_btn, upload_status],
    layout=Layout(align_items='center', gap='10px', margin='4px 0 0'),
)

# ── Drive folder source ─────────────────────────────────────────────────────
_DRIVE_FOLDER = '/content/drive/MyDrive/Inpainting-Test'
_AUDIO_EXTS   = {'.wav', '.mp3', '.flac', '.aiff', '.aif', '.ogg', '.m4a'}
_NO_FILES_OPT = '— no audio files found —'

def _scan_drive_folder():
    if not os.path.isdir(_DRIVE_FOLDER):
        return {}
    result = {}
    for fname in sorted(os.listdir(_DRIVE_FOLDER)):
        if os.path.splitext(fname)[1].lower() in _AUDIO_EXTS:
            result[fname] = os.path.join(_DRIVE_FOLDER, fname)
    return result

_drive_files = _scan_drive_folder()

_drive_ok = bool(_drive_files)
drive_dropdown = widgets.Dropdown(
    options=list(_drive_files) if _drive_ok else [_NO_FILES_OPT],
    description='',
    layout=Layout(width='360px'),
)
drive_load_btn = widgets.Button(
    description=' Load', icon='folder-open',
    layout=Layout(width='100px', height='34px'),
    disabled=not _drive_ok,
)
drive_load_btn.style.button_color = '#EDE9FE'
drive_refresh_btn = widgets.Button(
    description='', icon='refresh',
    layout=Layout(width='36px', height='34px'),
    tooltip='Re-scan folder',
)
_drive_status_val = (
    f'{len(_drive_files)} file(s) found'
    if _drive_ok
    else ('Drive not mounted or folder not found'
          if not os.path.isdir(_DRIVE_FOLDER)
          else 'No audio files in folder')
)
drive_status = widgets.HTML(
    value=f'<span style="font-size:12px;color:#9CA3AF;">{_drive_status_val}</span>',
)
drive_row = widgets.HBox(
    [drive_dropdown, drive_load_btn, drive_refresh_btn, drive_status],
    layout=Layout(align_items='center', gap='6px'),
)

waveform_out = widgets.Output()
audio_out    = widgets.Output()   # holds the persistent source audio player

zoom_slider = widgets.FloatRangeSlider(
    value=[0.0, 30.0], min=0.0, max=120.0, step=0.1,
    description='View (s):',
    style={'description_width': '60px'},
    layout=Layout(width='540px'),
    readout_format='.1f',
    continuous_update=False,
)

mask_start_box = widgets.BoundedFloatText(
    value=5.0, min=0.0, max=120.0, step=0.01,
    description='Start (s):',
    style={'description_width': '60px'},
    layout=Layout(width='175px'),
)
mask_end_box = widgets.BoundedFloatText(
    value=15.0, min=0.0, max=120.0, step=0.01,
    description='End (s):',
    style={'description_width': '52px'},
    layout=Layout(width='165px'),
)
mask_row = widgets.HBox(
    [mask_start_box, mask_end_box],
    layout=Layout(gap='16px', margin='2px 0'),
)

mask_label = widgets.HTML(
    value='<div class="sa3-stat" style="margin:2px 0;">No audio loaded yet.</div>'
)

preview_btn = widgets.Button(
    description='Preview mask', icon='headphones',
    layout=Layout(width='152px', height='36px'),
)
preview_btn.style.button_color = '#EDE9FE'
preview_out = widgets.Output()

inpaint_prompt = widgets.Textarea(
    value='Replace this region with a gentle acoustic guitar melody',
    placeholder='Describe what should replace the masked region\u2026',
    layout=Layout(width='100%', height='64px'),
    style={'description_width': '0px'},
)

inpaint_neg_prompt = widgets.Textarea(
    value='',
    placeholder='Optional negative prompt — what to avoid (e.g. “distortion, noise”)',
    layout=Layout(width='100%', height='44px'),
    style={'description_width': '0px'},
)

cfg_scale_slider = widgets.FloatSlider(
    value=1.0, min=0.0, max=15.0, step=0.5,
    description='',
    layout=Layout(width='490px'),
    readout_format='.1f',
    continuous_update=False,
)

cfg_guide = widgets.HTML(value=(
    '<div style="margin:4px 0 10px;font-family:-apple-system,BlinkMacSystemFont,\'Segoe UI\',sans-serif;">'
    '<p style="font-size:12px;color:#4B5563;margin:0 0 8px;line-height:1.6;">'
    'CFG (Classifier-Free Guidance) scale controls how strongly the model follows your '
    'prompt. It works by amplifying the difference between a prompt-conditioned and an '
    'unconditional prediction at each diffusion step &mdash; higher values push harder '
    'toward the prompt, at the cost of musical continuity with the surrounding audio.'
    '</p>'
    '<table style="width:100%;border-collapse:collapse;font-size:12px;">'
    '<thead><tr style="background:#EDE9FE;">'
    '<th style="text-align:left;padding:5px 10px;color:#5B21B6;font-weight:600;border-radius:4px 0 0 0;">CFG</th>'
    '<th style="text-align:left;padding:5px 10px;color:#5B21B6;font-weight:600;border-radius:0 4px 0 0;">What to expect</th>'
    '</tr></thead>'
    '<tbody>'
    '<tr style="border-bottom:1px solid #EDE9FE;"><td style="padding:5px 10px;color:#374151;font-weight:600;">0.0</td>'
    '<td style="padding:5px 10px;color:#374151;">Ignores the prompt entirely &mdash; fills the gap based on musical context alone. Good for &ldquo;fix a glitch&rdquo; or natural extension.</td></tr>'
    '<tr style="border-bottom:1px solid #EDE9FE;background:#FAFAFA;"><td style="padding:5px 10px;color:#374151;font-weight:600;">1.0 &#x2605;</td>'
    '<td style="padding:5px 10px;color:#374151;">Default. Mild prompt influence; continuity with the surrounding music is usually preserved.</td></tr>'
    '<tr style="border-bottom:1px solid #EDE9FE;"><td style="padding:5px 10px;color:#374151;font-weight:600;">3&ndash;5</td>'
    '<td style="padding:5px 10px;color:#374151;">Noticeably follows the prompt. Some risk of seam artifacts at the mask boundaries.</td></tr>'
    '<tr style="border-bottom:1px solid #EDE9FE;background:#FAFAFA;"><td style="padding:5px 10px;color:#374151;font-weight:600;">7&ndash;10</td>'
    '<td style="padding:5px 10px;color:#374151;">Strong adherence to the prompt. The replacement may feel disconnected from the surrounding music.</td></tr>'
    '<tr><td style="padding:5px 10px;color:#374151;font-weight:600;">12&ndash;15</td>'
    '<td style="padding:5px 10px;color:#374151;">Over-committed to the prompt. Often sounds harsh or incoherent.</td></tr>'
    '</tbody></table>'
    '<p style="font-size:11px;color:#9CA3AF;margin:6px 0 0;">'
    'Sweet spot for inpainting: <b style="color:#6B7280;">1.5&ndash;4.0</b> &mdash; enough to hear the prompt without breaking the musical seam.'
    '</p>'
    '</div>'
))

inpaint_steps_slider = widgets.IntSlider(
    value=8, min=4, max=50, step=1,
    description='', layout=Layout(width='490px'),
    continuous_update=False,
)
inpaint_steps_guide = widgets.HTML(value=(
    '<div style="font-size:12px;color:#4B5563;margin:2px 0 8px;line-height:1.6;">'
    'Denoising steps &mdash; the model is tuned for <b>8</b>. More steps refine texture '
    'with diminishing returns; cost grows roughly linearly (16 steps &asymp; 2&times; the time).'
    '</div>'
))

inpaint_seed = widgets.BoundedIntText(
    value=-1, min=-1, max=2**31-1,
    description='Seed:', style={'description_width': '40px'},
    layout=Layout(width='160px'),
)

inpaint_btn = widgets.Button(
    description='  Inpaint', icon='magic', button_style='',
    layout=Layout(width='150px', height='42px'),
)
inpaint_btn.style.button_color = '#7C3AED'
inpaint_btn.style.font_weight  = '700'

inpaint_out = widgets.Output(layout=Layout(min_height='60px', margin='6px 0 0'))


# ── Helpers ───────────────────────────────────────────────────────────────────
def _load_audio_into_state(audio_tensor, label):
    global _inpaint_audio
    _inpaint_audio = audio_tensor
    duration = audio_tensor.shape[-1] / SAMPLE_RATE
    ms = round(min(5.0,  duration * 0.15), 2)
    me = round(min(15.0, duration * 0.40), 2)
    # Update mask boxes
    mask_start_box.max   = round(duration, 2)
    mask_end_box.max     = round(duration, 2)
    mask_start_box.value = ms
    mask_end_box.value   = me
    # Update zoom slider to show full audio
    zoom_slider.max   = round(duration, 1)
    zoom_slider.value = [0.0, round(duration, 1)]
    upload_status.value = (
        f'<div class="sa3-info" style="margin:4px 0;">\u2705&nbsp; <strong>{label}</strong> &nbsp;|&nbsp; {duration:.1f}&thinsp;s &nbsp;|&nbsp; {audio_tensor.shape[0]} ch</div>'
    )
    _update_mask_label()
    _refresh_waveform()
    # Persist source audio player — save once, display once, no re-render
    _tmp_src = '/tmp/_sa3_source.wav'
    torchaudio.save(_tmp_src, audio_tensor.cpu(), SAMPLE_RATE)
    with audio_out:
        clear_output(wait=True)
        display(Audio(_tmp_src, autoplay=False))


def _refresh_waveform(*_):
    if _inpaint_audio is None:
        return
    start = mask_start_box.value
    end   = mask_end_box.value
    v0, v1 = zoom_slider.value
    png = _render_waveform(_inpaint_audio, start, end, v0, v1)
    with waveform_out:
        clear_output(wait=True)
        display(Image(data=png))


def _update_mask_label(*_):
    start, end = mask_start_box.value, mask_end_box.value
    mask_label.value = (
        f'<span class="sa3-stat">&#x23F1;&thinsp; '
        f'<strong>{start:.2f}s</strong> &ndash; <strong>{end:.2f}s</strong>'
        f' &nbsp;&nbsp; ({end - start:.2f}s)</span>'
    )


# ── Callbacks ─────────────────────────────────────────────────────────────────
def on_inpaint_model_change(change):
    new_model = change["new"]
    cfg = MODEL_CONFIGS[new_model]
    needs_warn = cfg["needs_gpu"] and not torch.cuda.is_available()
    inpaint_gpu_warning.layout.display = "" if needs_warn else "none"
    inpaint_limit_note.value = _limit_note_html(new_model)

inpaint_model_dropdown.observe(on_inpaint_model_change, names="value")


def on_source_change(change):
    title = change['new']
    if title == _UPLOAD_OPT:
        upload_row.layout.display = ''
    else:
        upload_row.layout.display = 'none'
        wav_path = _session_audio[title]
        audio, sr = torchaudio.load(wav_path)
        _load_audio_into_state(audio, title)


def on_upload(b):
    uploaded = colab_files.upload()
    if not uploaded:
        return
    for fname, data in uploaded.items():
        tmp = f'/tmp/{fname}'
        with open(tmp, 'wb') as f:
            f.write(data)
        audio, sr = torchaudio.load(tmp)
        if sr != SAMPLE_RATE:
            audio = torchaudio.functional.resample(audio, sr, SAMPLE_RATE)
        _load_audio_into_state(audio, fname)


def on_preview(b):
    if _inpaint_audio is None:
        return
    start = mask_start_box.value
    end   = mask_end_box.value
    if start >= end:
        with preview_out:
            clear_output(wait=True)
            display(HTML('<div class="sa3-error">Start must be less than End.</div>'))
        return
    segment = _inpaint_audio[:, int(start * SAMPLE_RATE):int(end * SAMPLE_RATE)]
    torchaudio.save('/tmp/_sa3_mask_preview.wav', segment.cpu(), SAMPLE_RATE)
    with preview_out:
        clear_output(wait=True)
        display(Audio('/tmp/_sa3_mask_preview.wav', autoplay=True))


def on_inpaint(b):
    if _inpaint_audio is None:
        with inpaint_out:
            clear_output(wait=True)
            display(HTML('<div class="sa3-error">Select or upload an audio file first.</div>'))
        return
    prompt = inpaint_prompt.value.strip()
    if not prompt:
        with inpaint_out:
            clear_output(wait=True)
            display(HTML('<div class="sa3-error">Please enter a prompt.</div>'))
        return

    model_name = inpaint_model_dropdown.value
    start = mask_start_box.value
    end   = mask_end_box.value
    if start >= end:
        with inpaint_out:
            clear_output(wait=True)
            display(HTML('<div class="sa3-error">Start must be less than End.</div>'))
        return
    duration = _inpaint_audio.shape[-1] / SAMPLE_RATE
    inpaint_btn.disabled    = True
    inpaint_btn.description = '  Inpainting\u2026'

    if model_name in _model_cache:
        _spin_status    = "Running inpainting diffusion\u2026"
        _spin_substatus = f"&#x26A1;&nbsp; <strong>{model_name}</strong> already loaded in memory."
    elif _is_model_on_disk(model_name):
        _spin_status    = f"Loading <strong>{model_name}</strong> from Drive cache\u2026"
        _spin_substatus = "&#x1F4BE;&nbsp; Weights found on Drive \u2014 no download needed."
    else:
        _spin_status    = f"Downloading <strong>{model_name}</strong> weights\u2026"
        _spin_substatus = "&#x2B07;&nbsp; First run \u2014 may take 1&ndash;3&thinsp;min."
    with inpaint_out:
        clear_output(wait=True)
        display(HTML(SPINNER_HTML.format(status=_spin_status, substatus=_spin_substatus)))

    try:
        import sys as _sys
        _rec = _sys.getrecursionlimit()
        _sys.setrecursionlimit(max(_rec, 4000))
        try:
            model = get_model(model_name)
        finally:
            _sys.setrecursionlimit(_rec)
        t0 = time.perf_counter()
        with torch.inference_mode():
            _generated = model.generate(
                prompt=prompt,
                negative_prompt=inpaint_neg_prompt.value.strip() or None,
                inpaint_audio=(SAMPLE_RATE, _inpaint_audio),
                inpaint_mask_start_seconds=start,
                inpaint_mask_end_seconds=end,
                duration=duration,
                steps=inpaint_steps_slider.value,
                cfg_scale=cfg_scale_slider.value,
                seed=inpaint_seed.value,
                chunked_decode=True,
            )
        elapsed   = time.perf_counter() - t0
        short     = prompt[:90] + ('…' if len(prompt) > 90 else '')
        s_samp       = int(start * SAMPLE_RATE)
        e_samp       = int(end   * SAMPLE_RATE)
        _seg_path    = '/tmp/_sa3_result_segment.wav'
        _full_path   = '/tmp/_sa3_result_full.wav'
        _clicks_path = '/tmp/_sa3_result_clicks.wav'
        import numpy as _np
        gen_audio = _generated[0].cpu()
        n_ch = gen_audio.shape[0]
        torchaudio.save(_seg_path,  gen_audio[:, s_samp:e_samp], SAMPLE_RATE)
        torchaudio.save(_full_path, gen_audio,                   SAMPLE_RATE)
        # Sample-accurate metronome clicks at mask start and end
        _click_dur = 0.012
        _click_n   = int(SAMPLE_RATE * _click_dur)
        _t   = _np.arange(_click_n) / SAMPLE_RATE
        _env = _np.exp(-_t / (_click_dur * 0.25))
        _wave = (_np.sin(2 * _np.pi * 1000.0 * _t) * _env * 0.7).astype(_np.float32)
        _click_t = torch.from_numpy(_wave).unsqueeze(0).expand(n_ch, -1).clone()
        clicked = gen_audio.clone()
        for _pos in [s_samp, e_samp]:
            _n = min(_click_n, clicked.shape[-1] - _pos)
            if _n > 0:
                clicked[:, _pos:_pos + _n] = (clicked[:, _pos:_pos + _n] + _click_t[:, :_n]).clamp(-1.0, 1.0)
        torchaudio.save(_clicks_path, clicked, SAMPLE_RATE)
        # Download widget
        _include_clicks_cb = widgets.Checkbox(
            value=True, description='Include clicks',
            indent=False, layout=Layout(width='auto'),
        )
        _dl_btn = widgets.Button(
            description='  Download', icon='download',
            layout=Layout(width='150px', height='34px'),
        )
        _dl_btn.style.button_color = '#EDE9FE'
        def _do_download(b, _fp=_full_path, _cp=_clicks_path):
            from google.colab import files as _cf
            _cf.download(_cp if _include_clicks_cb.value else _fp)
        _dl_btn.on_click(_do_download)
        _reg_title = _register_session_clip(f"Inpainted: {short[:40]}", _full_path)
        cap = MODEL_CONFIGS[model_name]['max_duration']
        with inpaint_out:
            clear_output(wait=True)
            display(HTML(f"""
            <div class=\"sa3-result\">
              <div style=\"display:flex;align-items:center;gap:8px;margin-bottom:8px;flex-wrap:wrap;\">
                <span class=\"sa3-badge\">{model_name}</span>
                <span class=\"sa3-stat\">mask {start:.2f}s–{end:.2f}s</span>
                <span class=\"sa3-stat\">CFG {cfg_scale_slider.value:.1f}</span>
                <span class=\"sa3-stat\">generated in {elapsed:.1f}&thinsp;s</span>
              </div>
              <div style=\"font-size:13px;color:#374151;font-style:italic;line-height:1.5;\">
                &ldquo;{short}&rdquo;
              </div>
            </div>
            """))
            display(HTML(f'<div style="font-size:12px;color:#6B7280;margin:6px 0 2px;">New segment ({end-start:.2f}&thinsp;s):</div>'))
            display(Audio(_seg_path, autoplay=True))
            display(HTML(f'<div style="font-size:12px;color:#6B7280;margin:8px 0 2px;">Model output ({cap:.0f}&thinsp;s) with click markers at mask boundaries:</div>'))
            display(Audio(_clicks_path, autoplay=False))
            display(widgets.HBox([_dl_btn, _include_clicks_cb], layout=Layout(align_items='center', gap='8px', margin='6px 0 0')))
            display(HTML(f'<div style="font-size:12px;color:#9CA3AF;margin:6px 0 0;">Saved to session clips as &ldquo;<b>{_reg_title}</b>&rdquo; &mdash; available as a source in every tool.</div>'))
    except RuntimeError as exc:
        with inpaint_out:
            clear_output(wait=True)
            display(HTML(f'<div class="sa3-error"><strong>Error:</strong> {exc}</div>'))
    finally:
        inpaint_btn.disabled    = False
        inpaint_btn.description = '  Inpaint'


def on_drive_load(b):
    fname = drive_dropdown.value
    if fname == _NO_FILES_OPT or fname not in _drive_files:
        return
    drive_status.value = '<span style="font-size:12px;color:#6B7280;">Loading…</span>'
    try:
        _audio, _sr = torchaudio.load(_drive_files[fname])
        if _sr != SAMPLE_RATE:
            _audio = torchaudio.functional.resample(_audio, _sr, SAMPLE_RATE)
        _load_audio_into_state(_audio, fname)
        drive_status.value = f'<span style="font-size:12px;color:#059669;">✓ {fname}</span>'
    except Exception as _exc:
        drive_status.value = f'<div class="sa3-error">Error loading: {_exc}</div>'

def on_drive_refresh(b):
    global _drive_files
    _drive_files = _scan_drive_folder()
    if _drive_files:
        drive_dropdown.options = list(_drive_files)
        drive_load_btn.disabled = False
        drive_status.value = f'<span style="font-size:12px;color:#6B7280;">{len(_drive_files)} file(s) found</span>'
    else:
        drive_dropdown.options = [_NO_FILES_OPT]
        drive_load_btn.disabled = True
        _msg = ('Drive not mounted or folder not found'
                if not os.path.isdir(_DRIVE_FOLDER)
                else 'No audio files in folder')
        drive_status.value = f'<span style="font-size:12px;color:#9CA3AF;">{_msg}</span>'

drive_load_btn.on_click(on_drive_load)
drive_refresh_btn.on_click(on_drive_refresh)

# ── Connect observers ─────────────────────────────────────────────────────────
source_dropdown.observe(on_source_change, names='value')
mask_start_box.observe(_update_mask_label, names='value')
mask_end_box.observe(_update_mask_label, names='value')
zoom_slider.observe(_refresh_waveform, names='value')
upload_btn.on_click(on_upload)
preview_btn.on_click(on_preview)
inpaint_btn.on_click(on_inpaint)

def _inpaint_refresh_sources():
    prev = source_dropdown.value
    opts = [_UPLOAD_OPT] + list(_session_audio.keys())
    source_dropdown.options = opts
    if prev in opts:
        source_dropdown.value = prev
_add_session_listener('inpaint', _inpaint_refresh_sources)


# ── Layout ────────────────────────────────────────────────────────────────────
# Pre-populate with any audio already saved before this cell ran
if _session_audio:
    source_dropdown.options = [_UPLOAD_OPT] + list(_session_audio.keys())

display(widgets.VBox([
    widgets.HTML('<div style="font-size:12px;color:#6B7280;margin:0 0 3px;">Model:</div>'),
    inpaint_model_dropdown,
    inpaint_gpu_warning,
    inpaint_limit_note,
    widgets.HTML('<div style="font-size:12px;color:#6B7280;margin:8px 0 3px;">Audio source:</div>'),
    source_dropdown,
    upload_row,
    widgets.HTML('<div style="font-size:12px;color:#6B7280;margin:6px 0 2px;">Drive folder — <code>My Drive/Inpainting-Test</code>:</div>'),
    drive_row,
    waveform_out,
    widgets.HTML('<div style="font-size:12px;color:#6B7280;margin:6px 0 2px;">Zoom (drag to set visible range):</div>'),
    zoom_slider,
    widgets.HTML('<div style="font-size:12px;color:#6B7280;margin:8px 0 2px;">Mask region to replace:</div>'),
    mask_row,
    mask_label,
    widgets.HBox([preview_btn, preview_out], layout=Layout(align_items='center', gap='12px')),
    widgets.HTML('<div style="font-size:12px;color:#6B7280;margin:6px 0 2px;">Source audio:</div>'),
    audio_out,
    widgets.HTML('<div style="font-size:12px;color:#6B7280;margin:10px 0 2px;">Prompt for the replacement region:</div>'),
    inpaint_prompt,
    widgets.HTML(
        '<div style="font-size:12px;color:#6B7280;margin:6px 0 2px;">'
        'Negative prompt <span style="color:#9CA3AF;">&mdash; optional</span></div>'
    ),
    inpaint_neg_prompt,
    widgets.HTML(
        '<div style="font-size:12px;color:#6B7280;margin:10px 0 1px;">'
        '<b>Prompt strength</b>&nbsp;'
        '<span style="color:#9CA3AF;">&mdash; how much to follow the prompt vs. blend with the surrounding music'
        ' (CFG scale, default 1.0)</span></div>'
        '<div style="display:flex;justify-content:space-between;width:490px;'
        'font-size:11px;color:#9CA3AF;margin:1px 0 2px;">'
        '<span>0 &mdash; blend with musical context</span>'
        '<span>15 &mdash; follow prompt strictly</span>'
        '</div>'
    ),
    cfg_scale_slider,
    cfg_guide,
    widgets.HTML(
        '<div style="font-size:12px;color:#6B7280;margin:8px 0 1px;">'
        '<b>Steps</b> '
        '<span style="color:#9CA3AF;">&mdash; denoising steps (default 8)</span></div>'
        '<div style="display:flex;justify-content:space-between;width:490px;'
        'font-size:11px;color:#9CA3AF;margin:1px 0 2px;">'
        '<span>4 &mdash; fastest</span><span>50 &mdash; max refinement</span></div>'
    ),
    inpaint_steps_slider,
    inpaint_steps_guide,
    inpaint_seed,
    inpaint_btn,
    inpaint_out,
], layout=Layout(width='100%', max_width='700px')))
```

```python
#@title 8 — RF-Inversion

display(HTML("""
<div class="sa3-section">
  <h2>8 &mdash; RF-Inversion</h2>
  <p>
    Re-style an audio clip without losing its structural identity. Instead of blending
    random noise into the audio (like audio-to-audio), RF-Inversion first runs a
    <em>forward ODE</em> to find the latent &ldquo;noise fingerprint&rdquo; that the
    model associates with your audio. Generation then denoises from <em>that</em>
    fingerprint guided by your new prompt, so the output shares the original&rsquo;s
    latent structure while adopting the prompt&rsquo;s character.
    Requires <strong>medium</strong> (GPU) for best results.
  </p>
</div>
"""))

# ── How it works (shown as a callout) ────────────────────────────────────────
display(HTML("""
<div class="sa3-card" style="margin-bottom:12px;">
  <strong style="font-size:13px;">How it differs from Audio-to-Audio</strong>
  <table style="width:100%;border-collapse:collapse;font-size:12px;margin-top:8px;">
    <thead><tr style="background:#EDE9FE;">
      <th style="padding:5px 10px;color:#5B21B6;text-align:left;">Mode</th>
      <th style="padding:5px 10px;color:#5B21B6;text-align:left;">Starting point</th>
      <th style="padding:5px 10px;color:#5B21B6;text-align:left;">What&rsquo;s preserved</th>
    </tr></thead>
    <tbody>
      <tr style="border-bottom:1px solid #EDE9FE;">
        <td style="padding:5px 10px;font-weight:600;">Audio-to-Audio</td>
        <td style="padding:5px 10px;color:#374151;">Audio latents + random noise</td>
        <td style="padding:5px 10px;color:#374151;">Musical context; timbre may drift</td>
      </tr>
      <tr style="background:#FAFAFA;">
        <td style="padding:5px 10px;font-weight:600;">RF-Inversion</td>
        <td style="padding:5px 10px;color:#374151;">Inverted latent (model&rsquo;s &ldquo;fingerprint&rdquo; of your audio)</td>
        <td style="padding:5px 10px;color:#374151;">Structural identity; re-style via prompt</td>
      </tr>
    </tbody>
  </table>
  <p style="font-size:11px;color:#9CA3AF;margin:8px 0 0;">
    Implementation: a &gamma;-controlled forward ODE finds the latent noise
    fingerprint (Phase&nbsp;1), then an &eta;-controlled reverse ODE denoises from it
    with your prompt while optionally pulling back toward the source (Phase&nbsp;2).
    Both phases run on the real rectified-flow model. If a future package version
    renames the internals, it falls back to guided audio-to-audio with a visible
    notice.
  </p>
</div>
"""))

# ── Imports ───────────────────────────────────────────────────────────────────
from google.colab import files as colab_files

# ── Constants ─────────────────────────────────────────────────────────────────
_RFI_DRIVE_FOLDER = '/content/drive/MyDrive/Inpainting-Test'
_RFI_AUDIO_EXTS   = {'.wav', '.mp3', '.flac', '.aiff', '.aif', '.ogg', '.m4a'}
_RFI_UPLOAD_OPT   = '— Upload new file —'
_RFI_NO_FILES_OPT = '— no audio files found —'

def _rfi_limit_html(model_name):
    cap = MODEL_CONFIGS[model_name]['max_duration']
    return (
        '<div style="font-size:12px;color:#4B5563;background:#F3F4F6;'
        'border-left:3px solid #9CA3AF;padding:6px 10px;border-radius:3px;margin:2px 0;">'
        f'ℹ️  <b>{model_name}</b> processes at most '
        f'<b>{cap:.0f} s</b> of the source.'
        '</div>'
    )

# ── State ─────────────────────────────────────────────────────────────────────
_rfi_audio = None

# ── Drive helpers ─────────────────────────────────────────────────────────────
def _rfi_scan_drive():
    if not os.path.isdir(_RFI_DRIVE_FOLDER):
        return {}
    return {
        fname: os.path.join(_RFI_DRIVE_FOLDER, fname)
        for fname in sorted(os.listdir(_RFI_DRIVE_FOLDER))
        if os.path.splitext(fname)[1].lower() in _RFI_AUDIO_EXTS
    }

_rfi_drive_files = _rfi_scan_drive()
_rfi_drive_ok    = bool(_rfi_drive_files)

# ── Core RF-Inversion implementation ─────────────────────────────────────────
def _rf_invert_and_generate(sa3, audio_tensor, prompt, neg_prompt,
                            gamma, n_steps, cfg_scale, duration,
                            eta=0.0, eta_stop=0.30):
    """
    True RF-Inversion (Rout et al., "Semantic Image Inversion and Editing using
    Rectified Stochastic Differential Equations"), implemented directly against
    the Stable Audio 3 rectified-flow model — no audio-to-audio substitution.

    Phase 1 — forward controlled ODE (t: 0 -> 1): integrate the source latent to
        a Gaussian "noise fingerprint" using the model's own velocity field
        blended (strength gamma) with a controller pulling toward a fixed target
        noise sample.
    Phase 2 — reverse controlled ODE (t: 1 -> 0): denoise from that fingerprint
        with the NEW prompt's velocity field (with CFG), blended (strength eta,
        active while t >= eta_stop) with a controller pulling back toward the
        original latent — this is what preserves structural identity.

    Conventions match the package's rectified flow exactly: t=1 is pure noise,
    t=0 is data, and the model returns velocity v with denoised = x - t*v. At
    gamma=0 the forward pass is the model's uncontrolled inversion; at eta=0 the
    reverse pass reduces *identically* to the stock Euler sampler used by
    model.generate(). Velocities come from sa3.dit (model.model) — the same
    object the library's samplers call — and conditioning, padding mask and the
    timestep schedule are built the same way generate() builds them.
    """
    from stable_audio_3.inference.audio_utils import prepare_audio
    from stable_audio_3.inference.sampling import build_schedule
    from stable_audio_3.data.utils import (
        compute_effective_seq_len_from_conditioning,
        create_padding_mask_from_lengths,
    )

    inner        = sa3.model                 # ConditionedDiffusionModelWrapper
    dit          = sa3.dit                   # DiTWrapper — velocity field the samplers call
    pretransform = inner.pretransform        # SAME autoencoder
    device       = sa3.device
    sr_model     = inner.sample_rate
    model_dtype  = next(dit.parameters()).dtype
    ae_dtype     = next(pretransform.parameters()).dtype
    ds_ratio     = pretransform.downsampling_ratio

    # ── Prepare + encode the source audio to a clean latent y0 ────────────────
    io_ch         = pretransform.io_channels
    audio_samples = int(duration * sr_model)
    audio_samples = ((audio_samples + ds_ratio - 1) // ds_ratio) * ds_ratio
    audio_p = prepare_audio(
        audio_tensor, in_sr=SAMPLE_RATE, target_sr=sr_model,
        target_length=audio_samples, target_channels=io_ch, device=str(device),
    ).to(ae_dtype)
    with torch.inference_mode():
        y0 = pretransform.encode(audio_p).to(torch.float32)   # (1, C_lat, T_lat)
    latent_len = y0.shape[-1]

    # ── Build conditioning EXACTLY like generate() (incl. empty inpaint chans) ─
    # generate() always injects zero inpaint_mask / inpaint_masked_input into the
    # input-concat channels; omitting them gives the DiT the wrong channel count.
    zero_mask  = torch.zeros((1, 1, latent_len), device=device, dtype=model_dtype)
    zero_input = torch.zeros((1, inner.io_channels, latent_len),
                             device=device, dtype=model_dtype)

    def _flatten(cond_list, negative):
        ct = inner.conditioner(cond_list, str(device))
        ct["inpaint_mask"]         = [zero_mask]
        ct["inpaint_masked_input"] = [zero_input]
        flat = inner.get_conditioning_inputs(ct, negative=negative)
        return {k: (v.type(model_dtype) if torch.is_tensor(v) else v)
                for k, v in flat.items()}

    edit_cond_list = [{"prompt": prompt or "", "seconds_total": duration}]
    null_inputs = _flatten([{"prompt": "", "seconds_total": duration}], negative=False)
    edit_inputs = _flatten(edit_cond_list, negative=False)
    if neg_prompt:
        edit_inputs = {**edit_inputs,
                       **_flatten([{"prompt": neg_prompt, "seconds_total": duration}],
                                  negative=True)}

    # ── Padding mask + timestep schedule (match sample_diffusion / generate) ──
    eff_len = compute_effective_seq_len_from_conditioning(
        edit_cond_list, sr_model, ds_ratio, str(device))
    padding_mask = None
    if eff_len is not None:
        headroom_tokens = int(6.0 * sr_model / ds_ratio)              # generate() default
        valid_lengths   = (eff_len + headroom_tokens).clamp(max=latent_len).long()
        padding_mask    = create_padding_mask_from_lengths(valid_lengths, latent_len)

    sigmas = build_schedule(
        steps=n_steps, sigma_max=1.0, dist_shift=inner.sampling_dist_shift,
        effective_seq_len=eff_len, fallback_seq_len=latent_len,
        include_endpoint=True, device=device,
    )
    if sigmas.dim() == 2:                       # per-element schedule -> batch==1 row
        sigmas = sigmas[0]
    sigmas = sigmas.to(torch.float32)           # descending: 1.0 -> 0.0

    def _velocity(x, t_scalar, inputs, cfg):
        # Same call the library samplers make on model.model, plus the generate()
        # CFG/APG flags. cfg=1.0 -> single conditional pass (returns velocity).
        t_eval = max(float(t_scalar), 1e-4)     # model is untrained at exactly t=0
        t_b = torch.full((x.shape[0],), t_eval, device=device, dtype=model_dtype)
        v = dit(x.to(model_dtype), t_b, cfg_scale=cfg, batch_cfg=True,
                rescale_cfg=True, padding_mask=padding_mask, apg_scale=1.0, **inputs)
        return v.to(torch.float32)

    # ── Phase 1 — forward controlled ODE (inversion), t: 0 -> 1 ───────────────
    y1_target = torch.randn_like(y0)            # fixed controller target noise
    asc = torch.flip(sigmas, dims=[0])          # ascending: 0.0 -> 1.0
    y = y0.clone()
    with torch.inference_mode():
        for i in range(asc.shape[0] - 1):
            t_cur = asc[i].item()
            dt    = (asc[i + 1] - asc[i]).item()                 # > 0
            v_uncond = _velocity(y, t_cur, null_inputs, 1.0)
            v_target = (y1_target - y) / max(1.0 - t_cur, 1e-2)  # toward noise endpoint
            v = (1.0 - gamma) * v_uncond + gamma * v_target
            y = y + dt * v
    inverted = y.detach()

    # ── Phase 2 — reverse controlled ODE (editing), t: 1 -> 0 ─────────────────
    x = inverted
    with torch.inference_mode():
        for i in range(sigmas.shape[0] - 1):
            t_cur = sigmas[i].item()
            dt    = (sigmas[i + 1] - sigmas[i]).item()           # < 0
            v_edit = _velocity(x, t_cur, edit_inputs, cfg_scale)
            use_eta = eta if (eta > 0.0 and t_cur >= eta_stop) else 0.0
            if use_eta > 0.0:
                v_struct = (x - y0) / max(t_cur, 1e-2)           # toward source latent
                v = (1.0 - use_eta) * v_edit + use_eta * v_struct
            else:
                v = v_edit
            x = x + dt * v

    # ── Decode latents -> audio ───────────────────────────────────────────────
    with torch.inference_mode():
        audio = pretransform.decode(x.to(ae_dtype), chunked=True)
    audio = audio.to(torch.float32).clamp(-1, 1)
    out = audio[0].cpu()
    target = audio_tensor.shape[-1]
    return out[:, :target] if out.shape[-1] > target else out


# ── Widgets ───────────────────────────────────────────────────────────────────
rfi_model_dd = widgets.Dropdown(
    options=[(MODEL_CONFIGS[k]["label"], k) for k in MODEL_CONFIGS],
    value="medium",
    description="Model:",
    style={"description_width": "52px"},
    layout=Layout(width="500px"),
)
rfi_gpu_warn = widgets.HTML(
    value='<div class="sa3-warn">⚠️  <b>medium</b> requires a GPU. '
          'Go to <b>Runtime → Change runtime type → GPU</b>.</div>',
    layout=Layout(display="" if MODEL_CONFIGS["medium"]["needs_gpu"] and not torch.cuda.is_available() else "none",
                  width="100%"),
)
rfi_limit_note = widgets.HTML(
    value=_rfi_limit_html("medium"), layout=Layout(width="100%")
)

# Audio source
rfi_source_dd = widgets.Dropdown(
    options=[_RFI_UPLOAD_OPT], description="Source:",
    style={"description_width": "60px"}, layout=Layout(width="440px"),
)
rfi_upload_btn = widgets.Button(
    description="Upload Audio", icon="upload",
    layout=Layout(width="155px", height="36px"),
)
rfi_upload_btn.style.button_color = "#EDE9FE"
rfi_upload_status = widgets.HTML(
    value='<div class="sa3-info" style="margin:4px 0;">No file loaded yet.</div>'
)
rfi_upload_row = widgets.HBox(
    [rfi_upload_btn, rfi_upload_status],
    layout=Layout(align_items="center", gap="10px", margin="4px 0 0"),
)

_rfi_dstat_val = (
    f'{len(_rfi_drive_files)} file(s) found' if _rfi_drive_ok
    else ('Drive not mounted or folder not found'
          if not os.path.isdir(_RFI_DRIVE_FOLDER) else 'No audio files in folder')
)
rfi_drive_dd = widgets.Dropdown(
    options=list(_rfi_drive_files) if _rfi_drive_ok else [_RFI_NO_FILES_OPT],
    description="", layout=Layout(width="360px"),
)
rfi_drive_load_btn = widgets.Button(
    description=" Load", icon="folder-open",
    layout=Layout(width="100px", height="34px"), disabled=not _rfi_drive_ok,
)
rfi_drive_load_btn.style.button_color = "#EDE9FE"
rfi_drive_refresh_btn = widgets.Button(
    description="", icon="refresh",
    layout=Layout(width="36px", height="34px"), tooltip="Re-scan folder",
)
rfi_drive_status = widgets.HTML(
    value=f'<span style="font-size:12px;color:#9CA3AF;">{_rfi_dstat_val}</span>'
)
rfi_drive_row = widgets.HBox(
    [rfi_drive_dd, rfi_drive_load_btn, rfi_drive_refresh_btn, rfi_drive_status],
    layout=Layout(align_items="center", gap="6px"),
)
rfi_audio_out = widgets.Output()

# Prompts
rfi_prompt = widgets.Textarea(
    value="",
    placeholder="Describe how to re-style the audio (e.g. “jazz piano trio, warm upright bass”)",
    layout=Layout(width="100%", height="56px"),
    style={"description_width": "0px"},
)
rfi_neg_prompt = widgets.Textarea(
    value="",
    placeholder="Optional negative prompt — what to avoid",
    layout=Layout(width="100%", height="44px"),
    style={"description_width": "0px"},
)

# Gamma
rfi_gamma_slider = widgets.FloatSlider(
    value=0.30, min=0.0, max=1.0, step=0.05,
    description="", layout=Layout(width="490px"),
    readout_format=".2f", continuous_update=False,
)
rfi_gamma_guide = widgets.HTML(value=(
    '<div style="margin:4px 0 10px;font-family:-apple-system,BlinkMacSystemFont,"Segoe UI",sans-serif;">'
    '<p style="font-size:12px;color:#4B5563;margin:0 0 8px;line-height:1.6;">'
    'Gamma controls how aggressively the forward ODE pushes the audio latent toward Gaussian noise. '
    'At <b>0</b> it only follows the model’s own unconditional drift, keeping the inverted latent '
    'close to the model’s semantic representation of the audio (best for style transfer). '
    'Higher values push harder toward pure noise, giving the generation phase more '
    'freedom — at the cost of fidelity to the original’s identity.'
    '</p>'
    '<table style="width:100%;border-collapse:collapse;font-size:12px;">'
    '<thead><tr style="background:#EDE9FE;">'
    '<th style="text-align:left;padding:5px 10px;color:#5B21B6;font-weight:600;">γ (Gamma)</th>'
    '<th style="text-align:left;padding:5px 10px;color:#5B21B6;font-weight:600;">Use case</th>'
    '</tr></thead><tbody>'
    '<tr style="border-bottom:1px solid #EDE9FE;">'
    '<td style="padding:5px 10px;font-weight:600;">0.0</td>'
    '<td style="padding:5px 10px;color:#374151;">Pure model drift — best starting point for style transfer with a strong prompt.</td></tr>'
    '<tr style="border-bottom:1px solid #EDE9FE;background:#FAFAFA;">'
    '<td style="padding:5px 10px;font-weight:600;">0.2–0.35 ★</td>'
    '<td style="padding:5px 10px;color:#374151;">Default. Balanced inversion; good mix of structural preservation and prompt responsiveness.</td></tr>'
    '<tr style="border-bottom:1px solid #EDE9FE;">'
    '<td style="padding:5px 10px;font-weight:600;">0.4–0.6</td>'
    '<td style="padding:5px 10px;color:#374151;">Stronger inversion toward noise. More generation freedom; recommended for faithful reconstruction.</td></tr>'
    '<tr>'
    '<td style="padding:5px 10px;font-weight:600;">0.7–1.0</td>'
    '<td style="padding:5px 10px;color:#374151;">Heavy noise injection. Output becomes loosely related to source; similar to audio-to-audio.</td></tr>'
    '</tbody></table></div>'
))

# Steps
rfi_steps_slider = widgets.IntSlider(
    value=50, min=10, max=150, step=5,
    description="", layout=Layout(width="490px"),
    continuous_update=False,
)
rfi_steps_guide = widgets.HTML(value=(
    '<div style="font-size:12px;color:#4B5563;margin:2px 0 8px;line-height:1.6;">'
    'Number of ODE steps for <em>both</em> the inversion and generation phases. '
    'More steps = higher quality inversion and smoother generation, at the cost of time. '
    '<b>50</b> is a good default; drop to <b>20–30</b> for fast previews.'
    '</div>'
))

# CFG
rfi_cfg_slider = widgets.FloatSlider(
    value=4.0, min=0.0, max=15.0, step=0.5,
    description="", layout=Layout(width="490px"),
    readout_format=".1f", continuous_update=False,
)
rfi_cfg_guide = widgets.HTML(value=(
    '<div style="font-size:12px;color:#4B5563;margin:2px 0 8px;line-height:1.6;">'
    'CFG scale for the generation phase. Because the inverted latent already encodes the '
    'original’s structure, higher CFG values (3–6) are safe and recommended to '
    'make the prompt’s re-style actually audible.'
    '</div>'
))

# Structure preservation (reverse controller eta)
rfi_eta_slider = widgets.FloatSlider(
    value=0.0, min=0.0, max=1.0, step=0.05,
    description="", layout=Layout(width="490px"),
    readout_format=".2f", continuous_update=False,
)
rfi_eta_stop_slider = widgets.FloatSlider(
    value=0.30, min=0.0, max=1.0, step=0.05,
    description="", layout=Layout(width="490px"),
    readout_format=".2f", continuous_update=False,
)
rfi_eta_guide = widgets.HTML(value=(
    '<div style="font-size:12px;color:#4B5563;margin:2px 0 8px;line-height:1.6;">'
    'During the reverse phase, how strongly to pull each step back toward the '
    '<em>source</em> latent. <b>0</b> = pure prompt-driven generation from the '
    'inverted fingerprint (maximum re-style). Raise toward <b>0.3\u20130.7</b> to '
    'lock the original\u2019s structure while still adopting the prompt. The '
    'controller is only active while the timestep is above the <b>release point</b> '
    'below, so late steps always follow the prompt \u2014 set the release point '
    'lower to hold structure longer.'
    '</div>'
))

# Seed
rfi_seed = widgets.BoundedIntText(
    value=-1, min=-1, max=2**31-1,
    description="Seed:", style={"description_width": "40px"},
    layout=Layout(width="160px"),
)

# Generate
rfi_btn = widgets.Button(
    description="  Run Inversion", icon="magic", button_style="",
    layout=Layout(width="180px", height="42px"),
)
rfi_btn.style.button_color = "#7C3AED"
rfi_btn.style.font_weight  = "700"
rfi_out = widgets.Output(layout=Layout(min_height="60px", margin="6px 0 0"))


# ── Helpers ───────────────────────────────────────────────────────────────────
def _rfi_load(audio_tensor, label):
    global _rfi_audio
    _rfi_audio = audio_tensor
    dur = audio_tensor.shape[-1] / SAMPLE_RATE
    rfi_upload_status.value = (
        f'<div class="sa3-info" style="margin:4px 0;">✅  '
        f'<strong>{label}</strong>  |  {dur:.1f} s '
        f' |  {audio_tensor.shape[0]} ch</div>'
    )
    _tmp = '/tmp/_sa3_rfi_src.wav'
    torchaudio.save(_tmp, audio_tensor.cpu(), SAMPLE_RATE)
    with rfi_audio_out:
        clear_output(wait=True)
        display(Audio(_tmp, autoplay=False))


# ── Callbacks ─────────────────────────────────────────────────────────────────
def _rfi_on_model(change):
    cfg = MODEL_CONFIGS[change["new"]]
    rfi_gpu_warn.layout.display = "" if cfg["needs_gpu"] and not torch.cuda.is_available() else "none"
    rfi_limit_note.value = _rfi_limit_html(change["new"])

def _rfi_on_source(change):
    title = change["new"]
    if title == _RFI_UPLOAD_OPT:
        rfi_upload_row.layout.display = ""
    else:
        rfi_upload_row.layout.display = "none"
        audio, sr = torchaudio.load(_session_audio[title])
        if sr != SAMPLE_RATE:
            audio = torchaudio.functional.resample(audio, sr, SAMPLE_RATE)
        _rfi_load(audio, title)

def _rfi_on_upload(b):
    uploaded = colab_files.upload()
    for fname, data in uploaded.items():
        tmp = f'/tmp/{fname}'
        with open(tmp, 'wb') as f:
            f.write(data)
        audio, sr = torchaudio.load(tmp)
        if sr != SAMPLE_RATE:
            audio = torchaudio.functional.resample(audio, sr, SAMPLE_RATE)
        _rfi_load(audio, fname)

def _rfi_on_drive_load(b):
    fname = rfi_drive_dd.value
    if fname not in _rfi_drive_files:
        return
    rfi_drive_status.value = '<span style="font-size:12px;color:#6B7280;">Loading…</span>'
    try:
        audio, sr = torchaudio.load(_rfi_drive_files[fname])
        if sr != SAMPLE_RATE:
            audio = torchaudio.functional.resample(audio, sr, SAMPLE_RATE)
        _rfi_load(audio, fname)
        rfi_drive_status.value = f'<span style="font-size:12px;color:#059669;">✓ {fname}</span>'
    except Exception as exc:
        rfi_drive_status.value = f'<div class="sa3-error">Error: {exc}</div>'

def _rfi_on_drive_refresh(b):
    global _rfi_drive_files, _rfi_drive_ok
    _rfi_drive_files = _rfi_scan_drive()
    _rfi_drive_ok = bool(_rfi_drive_files)
    if _rfi_drive_ok:
        rfi_drive_dd.options = list(_rfi_drive_files)
        rfi_drive_load_btn.disabled = False
        rfi_drive_status.value = f'<span style="font-size:12px;color:#6B7280;">{len(_rfi_drive_files)} file(s) found</span>'
    else:
        rfi_drive_dd.options = [_RFI_NO_FILES_OPT]
        rfi_drive_load_btn.disabled = True
        _msg = ("Drive not mounted or folder not found"
                if not os.path.isdir(_RFI_DRIVE_FOLDER) else "No audio files in folder")
        rfi_drive_status.value = f'<span style="font-size:12px;color:#9CA3AF;">{_msg}</span>'

def _rfi_on_run(b):
    if _rfi_audio is None:
        with rfi_out:
            clear_output(wait=True)
            display(HTML('<div class="sa3-error">Select or upload an audio file first.</div>'))
        return

    model_name = rfi_model_dd.value
    prompt     = rfi_prompt.value.strip() or None
    neg_prompt = rfi_neg_prompt.value.strip() or None
    gamma      = rfi_gamma_slider.value
    eta        = rfi_eta_slider.value
    eta_stop   = rfi_eta_stop_slider.value
    n_steps    = rfi_steps_slider.value
    cfg        = rfi_cfg_slider.value
    seed       = rfi_seed.value
    duration   = _rfi_audio.shape[-1] / SAMPLE_RATE

    rfi_btn.disabled    = True
    rfi_btn.description = "  Running…"

    if seed >= 0:
        torch.manual_seed(seed)
        if torch.cuda.is_available():
            torch.cuda.manual_seed_all(seed)

    if model_name in _model_cache:
        _st = "Phase 1: inverting audio…"
        _ss = f"⚡  <strong>{model_name}</strong> already loaded."
    elif _is_model_on_disk(model_name):
        _st = f"Loading <strong>{model_name}</strong>, then inverting…"
        _ss = "💾  Weights on Drive — no download needed."
    else:
        _st = f"Downloading <strong>{model_name}</strong>…"
        _ss = "⬇  First run — may take 1–3 min."

    with rfi_out:
        clear_output(wait=True)
        display(HTML(SPINNER_HTML.format(status=_st, substatus=_ss)))
    try:
        import sys as _sys
        _rec = _sys.getrecursionlimit()
        _sys.setrecursionlimit(max(_rec, 4000))
        try:
            sa3 = get_model(model_name)
        finally:
            _sys.setrecursionlimit(_rec)

        # ── Attempt true RF-Inversion ────────────────────────────────────────
        used_fallback = False
        t0 = time.perf_counter()
        try:
            with rfi_out:
                clear_output(wait=True)
                display(HTML(SPINNER_HTML.format(
                    status=f"Phase 1 of 2 — inverting ({n_steps} steps)…",
                    substatus="γ = {:.2f}  |  this may take a minute".format(gamma),
                )))
            result = _rf_invert_and_generate(
                sa3, _rfi_audio, prompt, neg_prompt, gamma, n_steps, cfg, duration,
                eta=eta, eta_stop=eta_stop,
            )
        except Exception as inv_exc:
            # Fall back to guided audio-to-audio
            used_fallback = True
            fallback_noise = min(0.35, gamma + 0.05)
            with rfi_out:
                clear_output(wait=True)
                display(HTML(SPINNER_HTML.format(
                    status="RF-Inversion unavailable — using guided A2A fallback…",
                    substatus=f"Reason: {inv_exc}",
                )))
            with torch.inference_mode():
                generated = sa3.generate(
                    prompt=prompt,
                    negative_prompt=neg_prompt,
                    init_audio=(SAMPLE_RATE, _rfi_audio),
                    init_noise_level=fallback_noise,
                    duration=duration,
                    steps=8,
                    cfg_scale=max(cfg, 3.0),
                    seed=seed,
                    chunked_decode=True,
                )
            result = generated[0].cpu()

        elapsed = time.perf_counter() - t0

        out_path = '/tmp/_sa3_rfi_result.wav'
        torchaudio.save(out_path, result, SAMPLE_RATE)
        _reg_title = _register_session_clip(f"Re-styled: {(prompt or 'untitled')[:40]}", out_path)

        _dl_btn = widgets.Button(
            description="  Download", icon="download",
            layout=Layout(width="150px", height="34px"),
        )
        _dl_btn.style.button_color = "#EDE9FE"
        def _do_dl(b, _p=out_path):
            from google.colab import files as _cf
            _cf.download(_p)
        _dl_btn.on_click(_do_dl)

        prompt_short = ((prompt or "")[:90] + "…") if prompt and len(prompt) > 90 else (prompt or "")
        mode_badge   = "A2A fallback" if used_fallback else "RF-Inversion"

        with rfi_out:
            clear_output(wait=True)
            if used_fallback:
                display(HTML(
                    '<div class="sa3-warn" style="margin-bottom:8px;">'
                    '⚠️  True RF-Inversion hit an unexpected error on this '
                    'package version and fell back to guided audio-to-audio. '
                    'The reason was shown above — please report it. '
                    'Results are similar for low gamma values.'
                    '</div>'
                ))
            display(HTML(f"""
            <div class="sa3-result">
              <div style="display:flex;align-items:center;gap:8px;margin-bottom:8px;flex-wrap:wrap;">
                <span class="sa3-badge">{model_name}</span>
                <span class="sa3-stat">{mode_badge}</span>
                <span class="sa3-stat">γ {gamma:.2f}</span>
                <span class="sa3-stat">CFG {cfg:.1f}</span>
                <span class="sa3-stat">{n_steps} steps</span>
                <span class="sa3-stat">done in {elapsed:.1f} s</span>
              </div>
              {'<div style="font-size:13px;color:#374151;font-style:italic;line-height:1.5;">&ldquo;' + prompt_short + '&rdquo;</div>' if prompt_short else ''}
            </div>
            """))
            display(HTML('<div style="font-size:12px;color:#6B7280;margin:6px 0 2px;">Result:</div>'))
            display(Audio(out_path, autoplay=True))
            display(HTML(f'<div style="font-size:12px;color:#9CA3AF;margin:4px 0 0;">Saved to session clips as &ldquo;<b>{_reg_title}</b>&rdquo; &mdash; available as a source in every tool.</div>'))
            display(_dl_btn)
    except Exception as exc:
        with rfi_out:
            clear_output(wait=True)
            display(HTML(f'<div class="sa3-error"><strong>Error:</strong> {exc}</div>'))
    finally:
        rfi_btn.disabled    = False
        rfi_btn.description = "  Run Inversion"


# ── Connect ───────────────────────────────────────────────────────────────────
rfi_model_dd.observe(_rfi_on_model, names="value")
rfi_source_dd.observe(_rfi_on_source, names="value")
rfi_upload_btn.on_click(_rfi_on_upload)
rfi_drive_load_btn.on_click(_rfi_on_drive_load)
rfi_drive_refresh_btn.on_click(_rfi_on_drive_refresh)
rfi_btn.on_click(_rfi_on_run)

def _rfi_refresh_sources():
    prev = rfi_source_dd.value
    opts = [_RFI_UPLOAD_OPT] + list(_session_audio.keys())
    rfi_source_dd.options = opts
    if prev in opts:
        rfi_source_dd.value = prev
_add_session_listener('rfi', _rfi_refresh_sources)
if _session_audio:
    rfi_source_dd.options = [_RFI_UPLOAD_OPT] + list(_session_audio.keys())

# ── Layout ────────────────────────────────────────────────────────────────────
display(widgets.VBox([
    widgets.HTML('<div style="font-size:12px;color:#6B7280;margin:0 0 3px;">Model:</div>'),
    rfi_model_dd,
    rfi_gpu_warn,
    rfi_limit_note,
    widgets.HTML('<div style="font-size:12px;color:#6B7280;margin:8px 0 3px;">Audio source:</div>'),
    rfi_source_dd,
    rfi_upload_row,
    widgets.HTML(
        f'<div style="font-size:12px;color:#6B7280;margin:6px 0 2px;">'
        f'Drive folder — <code>My Drive/{os.path.basename(_RFI_DRIVE_FOLDER)}</code>:</div>'
    ),
    rfi_drive_row,
    widgets.HTML('<div style="font-size:12px;color:#6B7280;margin:6px 0 2px;">Source audio:</div>'),
    rfi_audio_out,
    widgets.HTML(
        '<div style="font-size:12px;color:#6B7280;margin:10px 0 2px;">'
        'Re-style prompt <span style="color:#9CA3AF;">— describe the target sound character</span></div>'
    ),
    rfi_prompt,
    widgets.HTML(
        '<div style="font-size:12px;color:#6B7280;margin:6px 0 2px;">'
        'Negative prompt <span style="color:#9CA3AF;">— optional</span></div>'
    ),
    rfi_neg_prompt,
    widgets.HTML(
        '<div style="font-size:12px;color:#6B7280;margin:10px 0 1px;">'
        '<b>Gamma (γ)</b> '
        '<span style="color:#9CA3AF;">— inversion strength (default 0.30)</span></div>'
        '<div style="display:flex;justify-content:space-between;width:490px;'
        'font-size:11px;color:#9CA3AF;margin:1px 0 2px;">'
        '<span>0.0 — model drift only (style transfer)</span>'
        '<span>1.0 — full noise injection</span></div>'
    ),
    rfi_gamma_slider,
    rfi_gamma_guide,
    widgets.HTML(
        '<div style="font-size:12px;color:#6B7280;margin:8px 0 1px;">'
        '<b>ODE steps</b> '
        '<span style="color:#9CA3AF;">— used for both inversion and generation phases (default 50)</span></div>'
        '<div style="display:flex;justify-content:space-between;width:490px;'
        'font-size:11px;color:#9CA3AF;margin:1px 0 2px;">'
        '<span>10 — fast preview</span><span>150 — high quality</span></div>'
    ),
    rfi_steps_slider,
    rfi_steps_guide,
    widgets.HTML(
        '<div style="font-size:12px;color:#6B7280;margin:8px 0 1px;">'
        '<b>Prompt strength</b> '
        '<span style="color:#9CA3AF;">— CFG scale for generation phase (default 4.0)</span></div>'
        '<div style="display:flex;justify-content:space-between;width:490px;'
        'font-size:11px;color:#9CA3AF;margin:1px 0 2px;">'
        '<span>0 — ignore prompt</span><span>15 — follow strictly</span></div>'
    ),
    rfi_cfg_slider,
    rfi_cfg_guide,
    widgets.HTML(
        '<div style="font-size:12px;color:#6B7280;margin:8px 0 1px;">'
        '<b>Structure preservation</b> '
        '<span style="color:#9CA3AF;">&mdash; reverse controller \u03b7 (default 0.00 = off)</span></div>'
        '<div style="display:flex;justify-content:space-between;width:490px;'
        'font-size:11px;color:#9CA3AF;margin:1px 0 2px;">'
        '<span>0.00 \u2014 free re-style</span><span>1.00 \u2014 lock to source</span></div>'
    ),
    rfi_eta_slider,
    widgets.HTML(
        '<div style="font-size:12px;color:#6B7280;margin:6px 0 1px;">'
        '<b>Release point</b> '
        '<span style="color:#9CA3AF;">&mdash; stop the structure pull below this timestep (default 0.30)</span></div>'
    ),
    rfi_eta_stop_slider,
    rfi_eta_guide,
    rfi_seed,
    rfi_btn,
    rfi_out,
], layout=Layout(width="100%", max_width="700px")))
```

```python
#@title 9 — Audio Transition

display(HTML("""
<div class="sa3-section">
  <h2>9 &mdash; Audio Transition</h2>
  <p>
    Morph one clip into another with a smooth latent crossfade. Both clips are encoded
    by the SAME autoencoder, then across a transition region the latent frames are
    blended with a <strong>spherical interpolation (slerp)</strong> that respects the
    latent geometry. The result is decoded, then optionally pulled back onto the
    model&rsquo;s learned manifold with a low-noise <em>audio-to-audio</em> pass that
    smooths seam artifacts. The two creative dials are the <strong>transition length /
    curve</strong> and the <strong>re-coherence noise level</strong>. Uses
    <strong>medium</strong> (SAME-L, GPU) for best fidelity; <strong>small-music</strong>
    (SAME-S) works on constrained hardware at some quality cost.
  </p>
</div>
"""))

# ── Imports (other names come from cells 4–9) ─────────────────────────────────
from google.colab import files as colab_files

# ── Constants ─────────────────────────────────────────────────────────────────
_MORPH_DRIVE_FOLDER = '/content/drive/MyDrive/Inpainting-Test'  # edit to change
_MORPH_AUDIO_EXTS   = {'.wav', '.mp3', '.flac', '.aiff', '.aif', '.ogg', '.m4a'}
_MORPH_UPLOAD_OPT   = '— Upload new file —'
_MORPH_NO_FILES_OPT = '— no audio files found —'

# ── State: one slot per clip ──────────────────────────────────────────────────
_morph_audio = {'A': None, 'B': None}   # tag → Tensor (channels, samples) @ SAMPLE_RATE


# ── Blend curves ──────────────────────────────────────────────────────────────
def _morph_curve(p, kind):
    # p: Tensor in [0,1]. Returns blend parameter in [0,1].
    if kind == 'linear':
        return p
    if kind == 'ease-in-out':          # smoothstep
        return p * p * (3.0 - 2.0 * p)
    # 'steep' — smootherstep
    return p * p * p * (p * (p * 6.0 - 15.0) + 10.0)


# ── Per-frame spherical interpolation across the channel dim ──────────────────
def _morph_slerp(a, b, u, eps=1e-6):
    # a, b: (C, n) latent frames; u: (n,) blend per frame. Done in fp32 for stable acos.
    af = a.float(); bf = b.float(); uf = u.float()
    a_n = af / af.norm(dim=0, keepdim=True).clamp_min(eps)
    b_n = bf / bf.norm(dim=0, keepdim=True).clamp_min(eps)
    dot = (a_n * b_n).sum(0).clamp(-1.0 + 1e-6, 1.0 - 1e-6)   # (n,)
    omega = torch.acos(dot)
    so = torch.sin(omega)
    out = (torch.sin((1.0 - uf) * omega) / so) * af + (torch.sin(uf * omega) / so) * bf
    lin = (1.0 - uf) * af + uf * bf                           # near-collinear fallback
    blended = torch.where(so.abs() > 1e-4, out, lin)
    return blended.to(a.dtype)


# ── Core morph: encode → slerp crossfade → decode ─────────────────────────────
def _morph_build_latent(dm, clip_a, clip_b, transition_seconds, curve_kind):
    from stable_audio_3.inference.audio_utils import prepare_audio

    ae       = dm.pretransform
    device   = next(dm.parameters()).device
    dtype    = next(dm.parameters()).dtype
    sr_model = dm.sample_rate
    io_ch    = ae.io_channels if ae is not None else dm.io_channels

    def _encode(clip):
        dur = clip.shape[-1] / SAMPLE_RATE
        audio_p = prepare_audio(
            clip, in_sr=SAMPLE_RATE, target_sr=sr_model,
            target_length=int(dur * sr_model), target_channels=io_ch,
            device=str(device),
        ).to(dtype)
        with torch.inference_mode():
            lat = ae.encode(audio_p)        # (1, C_lat, T_lat)
        return lat, dur

    lat_a, dur_a = _encode(clip_a)
    lat_b, dur_b = _encode(clip_b)

    t_a = lat_a.shape[-1]
    t_b = lat_b.shape[-1]
    fps = t_a / dur_a                        # latent frames per second (~10.76)

    # Transition length in frames, clamped to what both clips can supply.
    n = int(round(transition_seconds * fps))
    n = max(1, min(n, t_a, t_b))

    # Blend parameter per transition frame, shaped by the chosen curve.
    if n >= 2:
        p = torch.linspace(0.0, 1.0, n, device=device)
    else:
        p = torch.tensor([0.5], device=device)
    u = _morph_curve(p, curve_kind)          # (n,)

    a_block = lat_a[0, :, t_a - n:]          # (C, n) — A's tail
    b_block = lat_b[0, :, :n]                # (C, n) — B's head
    trans   = _morph_slerp(a_block, b_block, u).unsqueeze(0)   # (1, C, n)

    head = lat_a[..., : t_a - n]             # pure A
    tail = lat_b[..., n:]                    # pure B
    morph_lat = torch.cat([head, trans, tail], dim=-1)

    with torch.inference_mode():
        raw = ae.decode(morph_lat)           # (1, C, T) — chunked decode off
    raw = raw[0].cpu()

    info = {'fps': fps, 'n_frames': n, 'trans_sec': n / fps,
            't_a': t_a, 't_b': t_b, 'out_frames': morph_lat.shape[-1]}
    return raw, info


# ── Drive helpers ─────────────────────────────────────────────────────────────
def _morph_scan_drive():
    if not os.path.isdir(_MORPH_DRIVE_FOLDER):
        return {}
    return {
        fname: os.path.join(_MORPH_DRIVE_FOLDER, fname)
        for fname in sorted(os.listdir(_MORPH_DRIVE_FOLDER))
        if os.path.splitext(fname)[1].lower() in _MORPH_AUDIO_EXTS
    }

_morph_drive_files = _morph_scan_drive()
_morph_drive_ok    = bool(_morph_drive_files)


def _morph_frames_readout():
    a, b = _morph_audio['A'], _morph_audio['B']
    if a is None or b is None:
        return ('<span style="font-size:12px;color:#9CA3AF;">'
                'Load both clips to see the transition frame count.</span>')
    sec = morph_trans_slider.value
    n_est = max(1, int(round(sec * 10.76)))
    return (f'<span style="font-size:12px;color:#6B7280;">A {a.shape[-1]/SAMPLE_RATE:.1f}s · '
            f'B {b.shape[-1]/SAMPLE_RATE:.1f}s · transition ≈ <b>{n_est}</b> latent '
            f'frames (~{sec:.1f}s @ ~10.76 fps)</span>')


# ── Source-selector factory (one per clip, mirrors the A2A pattern) ───────────
def _morph_make_source(tag):
    src_dd = widgets.Dropdown(
        options=[_MORPH_UPLOAD_OPT], description=f"Clip {tag}:",
        style={"description_width": "50px"}, layout=Layout(width="430px"),
    )
    upload_btn = widgets.Button(
        description="Upload", icon="upload", layout=Layout(width="120px", height="34px"),
    )
    upload_btn.style.button_color = "#EDE9FE"
    status = widgets.HTML(
        value='<div class="sa3-info" style="margin:4px 0;">No file loaded yet.</div>'
    )
    upload_row = widgets.HBox(
        [upload_btn, status], layout=Layout(align_items="center", gap="10px", margin="4px 0 0"),
    )
    drive_dd = widgets.Dropdown(
        options=list(_morph_drive_files) if _morph_drive_ok else [_MORPH_NO_FILES_OPT],
        description="", layout=Layout(width="340px"),
    )
    drive_load_btn = widgets.Button(
        description=" Load", icon="folder-open",
        layout=Layout(width="95px", height="32px"), disabled=not _morph_drive_ok,
    )
    drive_load_btn.style.button_color = "#EDE9FE"
    _dstat = (f'{len(_morph_drive_files)} file(s) found' if _morph_drive_ok
              else 'no Drive files')
    drive_status = widgets.HTML(
        value=f'<span style="font-size:12px;color:#9CA3AF;">{_dstat}</span>'
    )
    drive_row = widgets.HBox(
        [drive_dd, drive_load_btn, drive_status],
        layout=Layout(align_items="center", gap="6px"),
    )
    audio_out = widgets.Output()

    def _load(audio_tensor, label):
        _morph_audio[tag] = audio_tensor
        dur = audio_tensor.shape[-1] / SAMPLE_RATE
        status.value = (
            f'<div class="sa3-info" style="margin:4px 0;">✅  '
            f'<strong>{label}</strong>  |  {dur:.1f} s  |  {audio_tensor.shape[0]} ch</div>'
        )
        _tmp = f'/tmp/_sa3_morph_src_{tag}.wav'
        torchaudio.save(_tmp, audio_tensor.cpu(), SAMPLE_RATE)
        with audio_out:
            clear_output(wait=True)
            display(Audio(_tmp, autoplay=False))
        morph_frames_note.value = _morph_frames_readout()
        _morph_update_trans_max()

    def _on_source(change):
        title = change["new"]
        if title == _MORPH_UPLOAD_OPT:
            upload_row.layout.display = ""
        else:
            upload_row.layout.display = "none"
            audio, sr = torchaudio.load(_session_audio[title])
            if sr != SAMPLE_RATE:
                audio = torchaudio.functional.resample(audio, sr, SAMPLE_RATE)
            _load(audio, title)

    def _on_upload(b):
        uploaded = colab_files.upload()
        for fname, data in uploaded.items():
            tmp = f'/tmp/{fname}'
            with open(tmp, 'wb') as f:
                f.write(data)
            audio, sr = torchaudio.load(tmp)
            if sr != SAMPLE_RATE:
                audio = torchaudio.functional.resample(audio, sr, SAMPLE_RATE)
            _load(audio, fname)

    def _on_drive_load(b):
        fname = drive_dd.value
        if fname not in _morph_drive_files:
            return
        drive_status.value = '<span style="font-size:12px;color:#6B7280;">Loading…</span>'
        try:
            audio, sr = torchaudio.load(_morph_drive_files[fname])
            if sr != SAMPLE_RATE:
                audio = torchaudio.functional.resample(audio, sr, SAMPLE_RATE)
            _load(audio, fname)
            drive_status.value = f'<span style="font-size:12px;color:#059669;">✓ {fname}</span>'
        except Exception as exc:
            drive_status.value = f'<div class="sa3-error">Error: {exc}</div>'

    src_dd.observe(_on_source, names="value")
    upload_btn.on_click(_on_upload)
    drive_load_btn.on_click(_on_drive_load)
    def _refresh_sources():
        prev = src_dd.value
        opts = [_MORPH_UPLOAD_OPT] + list(_session_audio.keys())
        src_dd.options = opts
        if prev in opts:
            src_dd.value = prev
    _add_session_listener(f'morph-{tag}', _refresh_sources)
    if _session_audio:
        src_dd.options = [_MORPH_UPLOAD_OPT] + list(_session_audio.keys())

    box = widgets.VBox([
        src_dd, upload_row,
        widgets.HTML(
            f'<div style="font-size:12px;color:#6B7280;margin:4px 0 2px;">'
            f'Drive — <code>My Drive/{os.path.basename(_MORPH_DRIVE_FOLDER)}</code>:</div>'
        ),
        drive_row, audio_out,
    ])
    return box


# ── Shared widgets ────────────────────────────────────────────────────────────
morph_model_dd = widgets.Dropdown(
    options=[(MODEL_CONFIGS[k]["label"], k) for k in MODEL_CONFIGS],
    value="medium", description="Model:",
    style={"description_width": "52px"}, layout=Layout(width="500px"),
)
morph_gpu_warn = widgets.HTML(
    value='<div class="sa3-warn">⚠️  <b>medium</b> requires a GPU with Flash Attention. '
          'Go to <b>Runtime → Change runtime type → GPU</b>, or switch to '
          '<b>small-music</b> above.</div>',
    layout=Layout(display="" if MODEL_CONFIGS["medium"]["needs_gpu"] and not torch.cuda.is_available() else "none",
                  width="100%"),
)

morph_src_a = _morph_make_source('A')
morph_src_b = _morph_make_source('B')

morph_trans_slider = widgets.FloatSlider(
    value=3.0, min=0.5, max=60.0, step=0.5,
    description="", layout=Layout(width="490px"),
    readout_format=".1f", continuous_update=False,
)

def _morph_update_trans_max():
    a, b = _morph_audio['A'], _morph_audio['B']
    if a is not None and b is not None:
        total_duration = (a.shape[-1] + b.shape[-1]) / SAMPLE_RATE
        morph_trans_slider.max = max(60.0, total_duration)
    else:
        morph_trans_slider.max = 60.0
morph_curve_dd = widgets.Dropdown(
    options=[("Ease-in-out (smooth)", "ease-in-out"),
             ("Linear (constant)", "linear"),
             ("Steep (decisive)", "steep")],
    value="ease-in-out", description="",
    layout=Layout(width="280px"),
)
morph_frames_note = widgets.HTML(value=_morph_frames_readout())

morph_prompt = widgets.Textarea(
    value="",
    placeholder="Optional — describe the hybrid target for the re-coherence pass "
                "(e.g. “warm hybrid of both clips, seamless blend”). Blank = free re-cohere.",
    layout=Layout(width="100%", height="52px"), style={"description_width": "0px"},
)
morph_neg_prompt = widgets.Textarea(
    value="",
    placeholder="Optional negative prompt — what to avoid (e.g. “clicks, seam artifacts”)",
    layout=Layout(width="100%", height="40px"), style={"description_width": "0px"},
)

morph_noise_slider = widgets.FloatSlider(
    value=0.30, min=0.0, max=1.0, step=0.02,
    description="", layout=Layout(width="490px"),
    readout_format=".2f", continuous_update=False,
)
morph_noise_guide = widgets.HTML(value=(
    '<div style="font-size:12px;color:#4B5563;margin:4px 0 8px;line-height:1.6;">'
    'The re-coherence pass feeds the raw morph back through audio-to-audio to legalize '
    'seams onto the model&rsquo;s manifold. <b>0.2–0.4</b> is the sweet spot: too low and '
    'seam artifacts survive, too high and you lose the morph structure. '
    '<b>0.0</b> skips re-coherence and returns the raw slerp decode only — useful for '
    'judging whether the SAME midpoint is already musical.</div>'
))

morph_use_rfi = widgets.Checkbox(
    value=False,
    description="Re-cohere with RF-Inversion instead of blind low-noise A2A",
    indent=False,
    layout=Layout(width="100%"),
)
morph_rfi_note = widgets.HTML(value=(
    '<div style="font-size:11px;color:#9CA3AF;margin:0 0 6px;line-height:1.5;">'
    'Off (default): seam repair via low-noise audio-to-audio. On: invert the raw '
    'morph and regenerate it on the model\u2019s manifold (run section 8 once first). '
    'The re-coherence noise above is reused as the inversion strength.'
    '</div>'
))

morph_cfg_slider = widgets.FloatSlider(
    value=1.0, min=0.0, max=8.0, step=0.05,
    description="", layout=Layout(width="490px"),
    readout_format=".2f", continuous_update=False,
)
morph_seed = widgets.BoundedIntText(
    value=-1, min=-1, max=2**31-1,
    description="Seed:", style={"description_width": "40px"}, layout=Layout(width="160px"),
)

morph_btn = widgets.Button(
    description="  Create Transition", icon="magic",
    layout=Layout(width="200px", height="42px"),
)
morph_btn.style.button_color = "#7C3AED"
morph_btn.style.font_weight  = "700"
morph_out = widgets.Output(layout=Layout(min_height="60px", margin="6px 0 0"))


# ── Callbacks ─────────────────────────────────────────────────────────────────
def _morph_on_model(change):
    cfg = MODEL_CONFIGS[change["new"]]
    morph_gpu_warn.layout.display = "" if cfg["needs_gpu"] and not torch.cuda.is_available() else "none"

def _morph_on_trans(change):
    morph_frames_note.value = _morph_frames_readout()


def _morph_dl_button(path, label="  Download"):
    btn = widgets.Button(description=label, icon="download",
                         layout=Layout(width="150px", height="34px"))
    btn.style.button_color = "#EDE9FE"
    def _do(b, _p=path):
        from google.colab import files as _cf
        _cf.download(_p)
    btn.on_click(_do)
    return btn


def _morph_on_run(b):
    if _morph_audio['A'] is None or _morph_audio['B'] is None:
        with morph_out:
            clear_output(wait=True)
            display(HTML('<div class="sa3-error">Load both Clip A and Clip B first.</div>'))
        return

    model_name = morph_model_dd.value
    prompt     = morph_prompt.value.strip() or None
    neg_prompt = morph_neg_prompt.value.strip() or None
    trans_sec  = morph_trans_slider.value
    curve_kind = morph_curve_dd.value
    noise      = morph_noise_slider.value
    cfg        = morph_cfg_slider.value
    seed       = morph_seed.value

    morph_btn.disabled    = True
    morph_btn.description = "  Working…"

    if model_name in _model_cache:
        _st, _ss = "Encoding clips…", f"⚡  <strong>{model_name}</strong> already loaded."
    elif _is_model_on_disk(model_name):
        _st, _ss = f"Loading <strong>{model_name}</strong>…", "💾  Weights on Drive — no download needed."
    else:
        _st, _ss = f"Downloading <strong>{model_name}</strong>…", "⬇  First run — may take 1–3 min."

    with morph_out:
        clear_output(wait=True)
        display(HTML(SPINNER_HTML.format(status=_st, substatus=_ss)))
    try:
        import sys as _sys
        _rec = _sys.getrecursionlimit()
        _sys.setrecursionlimit(max(_rec, 4000))
        try:
            sa3 = get_model(model_name)
        finally:
            _sys.setrecursionlimit(_rec)

        # If cached on CPU but GPU is now available, reload on CUDA (matches A2A/RFI).
        import gc as _gc2
        _want = "cuda" if torch.cuda.is_available() else "cpu"
        _have = str(next(sa3.model.parameters()).device)
        if not _have.startswith(_want):
            del _model_cache[model_name]
            _gc2.collect()
            if torch.cuda.is_available():
                torch.cuda.empty_cache()
            _sys.setrecursionlimit(max(_sys.getrecursionlimit(), 4000))
            sa3 = get_model(model_name)
            _have = str(next(sa3.model.parameters()).device)
        _dev_label = "GPU ✓" if _have.startswith("cuda") else "CPU"

        if seed >= 0:
            torch.manual_seed(seed)
            if torch.cuda.is_available():
                torch.cuda.manual_seed_all(seed)

        # ── Stage 1: encode → slerp crossfade → decode ────────────────────────
        with morph_out:
            clear_output(wait=True)
            display(HTML(SPINNER_HTML.format(
                status="Building latent crossfade (slerp)…",
                substatus=f"transition ≈ {trans_sec:.1f}s · {curve_kind} curve",
            )))
        t0 = time.perf_counter()
        raw, info = _morph_build_latent(
            sa3.model, _morph_audio['A'], _morph_audio['B'], trans_sec, curve_kind,
        )
        raw_path = '/tmp/_sa3_morph_raw.wav'
        torchaudio.save(raw_path, raw, SAMPLE_RATE)

        # ── Stage 2: re-coherence — blind low-noise A2A (default) or RF-Inversion ─
        cohered_path = None
        _method_label = "raw decode"
        if noise > 0.0:
            use_rfi = morph_use_rfi.value
            if use_rfi and "_rf_invert_and_generate" not in globals():
                raise RuntimeError(
                    "RF-Inversion re-coherence needs section 8 (RF-Inversion) to have "
                    "been run at least once this session so its helper is defined."
                )
            _method_label = "RF-Inversion" if use_rfi else "A2A re-cohere"
            with morph_out:
                clear_output(wait=True)
                display(HTML(SPINNER_HTML.format(
                    status=("Re-cohering (RF-Inversion)…" if use_rfi
                            else "Re-cohering (low-noise audio-to-audio)…"),
                    substatus=f"noise = {noise:.2f}",
                )))
            with torch.inference_mode():
                if use_rfi:
                    # Reuse the re-coherence noise as the forward-inversion strength
                    # (gamma); a moderate reverse controller (eta) keeps the morph's
                    # structure while regenerating it on the model's manifold.
                    cohered = _rf_invert_and_generate(
                        sa3, raw, prompt, neg_prompt,
                        gamma=noise, n_steps=50, cfg_scale=cfg,
                        duration=raw.shape[-1] / SAMPLE_RATE,
                        eta=0.5, eta_stop=0.30,
                    )
                else:
                    generated = sa3.generate(
                        prompt=prompt, negative_prompt=neg_prompt,
                        init_audio=(SAMPLE_RATE, raw), init_noise_level=noise,
                        duration=raw.shape[-1] / SAMPLE_RATE,
                        steps=8, cfg_scale=cfg, seed=seed, chunked_decode=True,
                    )
                    cohered = generated[0].cpu()
            cohered_path = '/tmp/_sa3_morph_cohered.wav'
            torchaudio.save(cohered_path, cohered, SAMPLE_RATE)
        elapsed = time.perf_counter() - t0
        _best_path = cohered_path if cohered_path is not None else raw_path
        _reg_title = _register_session_clip("Morph A\u2192B", _best_path)

        with morph_out:
            clear_output(wait=True)
            display(HTML(f"""
            <div class="sa3-result">
              <div style="display:flex;align-items:center;gap:8px;margin-bottom:8px;flex-wrap:wrap;">
                <span class="sa3-badge">{model_name}</span>
                <span class="sa3-stat">{_dev_label}</span>
                <span class="sa3-stat">transition {info['trans_sec']:.1f}s · {info['n_frames']} frames</span>
                <span class="sa3-stat">{curve_kind}</span>
                <span class="sa3-stat">noise {noise:.2f}</span>
                <span class="sa3-stat">{_method_label}</span>
                <span class="sa3-stat">done in {elapsed:.1f} s</span>
              </div>
            </div>
            """))
            display(HTML(f'<div style="font-size:12px;color:#9CA3AF;margin:0 0 4px;">Saved to session clips as &ldquo;<b>{_reg_title}</b>&rdquo; &mdash; available as a source in every tool.</div>'))
            # Raw slerp decode — the diagnostic listen
            display(HTML(
                '<div style="font-size:12px;color:#6B7280;margin:8px 0 2px;">'
                '<b>Raw slerp decode</b> — before repair (diagnostic):</div>'
            ))
            display(Audio(raw_path, autoplay=False))
            display(_morph_dl_button(raw_path, "  Download raw"))
            # Re-cohered result
            if cohered_path is not None:
                display(HTML(
                    '<div style="font-size:12px;color:#6B7280;margin:12px 0 2px;">'
                    '<b>Re-cohered result</b> — low-noise audio-to-audio repair:</div>'
                ))
                display(Audio(cohered_path, autoplay=True))
                display(_morph_dl_button(cohered_path, "  Download result"))
            else:
                display(HTML(
                    '<div class="sa3-info" style="margin:10px 0 0;">'
                    'Re-coherence skipped (noise = 0). The raw decode above is the output.</div>'
                ))
    except Exception as exc:
        with morph_out:
            clear_output(wait=True)
            display(HTML(f'<div class="sa3-error"><strong>Error:</strong> {exc}</div>'))
    finally:
        morph_btn.disabled    = False
        morph_btn.description = "  Create Transition"


# ── Connect ───────────────────────────────────────────────────────────────────
morph_model_dd.observe(_morph_on_model, names="value")
morph_trans_slider.observe(_morph_on_trans, names="value")
morph_btn.on_click(_morph_on_run)


# ── Layout ────────────────────────────────────────────────────────────────────
display(widgets.VBox([
    widgets.HTML('<div style="font-size:12px;color:#6B7280;margin:0 0 3px;">Model:</div>'),
    morph_model_dd,
    morph_gpu_warn,
    widgets.HTML('<div style="font-size:13px;color:#5B21B6;font-weight:600;margin:10px 0 3px;">'
                 'Clip A — start</div>'),
    morph_src_a,
    widgets.HTML('<div style="font-size:13px;color:#5B21B6;font-weight:600;margin:12px 0 3px;">'
                 'Clip B — end</div>'),
    morph_src_b,
    widgets.HTML(
        '<div style="font-size:12px;color:#6B7280;margin:14px 0 1px;">'
        '<b>Transition length</b> '
        '<span style="color:#9CA3AF;">— seconds of A→B crossfade (default 3.0)</span></div>'
    ),
    morph_trans_slider,
    widgets.HTML('<div style="font-size:12px;color:#6B7280;margin:8px 0 1px;">'
                 '<b>Blend curve</b> '
                 '<span style="color:#9CA3AF;">— gradual vs decisive morph</span></div>'),
    morph_curve_dd,
    morph_frames_note,
    widgets.HTML(
        '<div style="font-size:12px;color:#6B7280;margin:12px 0 2px;">'
        'Re-coherence prompt <span style="color:#9CA3AF;">— optional; describes the hybrid target</span></div>'
    ),
    morph_prompt,
    widgets.HTML(
        '<div style="font-size:12px;color:#6B7280;margin:6px 0 2px;">'
        'Negative prompt <span style="color:#9CA3AF;">— optional</span></div>'
    ),
    morph_neg_prompt,
    widgets.HTML(
        '<div style="font-size:12px;color:#6B7280;margin:12px 0 1px;">'
        '<b>Re-coherence noise</b> '
        '<span style="color:#9CA3AF;">— manifold repair strength (0 = skip, default 0.30)</span></div>'
        '<div style="display:flex;justify-content:space-between;width:490px;'
        'font-size:11px;color:#9CA3AF;margin:1px 0 2px;">'
        '<span>0.0 — raw decode only</span><span>1.0 — heavy repair</span></div>'
    ),
    morph_noise_slider,
    morph_noise_guide,
    morph_use_rfi,
    morph_rfi_note,
    widgets.HTML(
        '<div style="font-size:12px;color:#6B7280;margin:8px 0 1px;">'
        '<b>Prompt strength</b> '
        '<span style="color:#9CA3AF;">— CFG for the re-coherence pass (ignored without a prompt)</span></div>'
    ),
    morph_cfg_slider,
    morph_seed,
    morph_btn,
    morph_out,
], layout=Layout(width="100%", max_width="720px")))
```

```python
#@title 10 — Extend & Loop

display(HTML("""
<div class="sa3-section">
  <h2>10 &mdash; Extend &amp; Loop</h2>
  <p>
    Three outpainting tools built on the same inpainting engine as section&nbsp;7 —
    the model regenerates only a chosen region and everything else is preserved exactly.
    <strong>Continue end</strong> writes new material past the end of a clip;
    <strong>Extend intro</strong> composes a lead-in before it;
    <strong>Seamless loop</strong> rotates the clip so its start/end joint sits in the
    middle, regenerates the joint, and rotates back &mdash; the result loops cleanly.
    Works with both <strong>small-music</strong> (CPU-ok) and <strong>medium</strong>.
  </p>
</div>
"""))

# ── Imports (other names come from cells 4–5) ──────────────────────────────────
from google.colab import files as colab_files

# ── Constants ─────────────────────────────────────────────────────────────────
_EXT_DRIVE_FOLDER = '/content/drive/MyDrive/Inpainting-Test'  # edit to change
_EXT_AUDIO_EXTS   = {'.wav', '.mp3', '.flac', '.aiff', '.aif', '.ogg', '.m4a'}
_EXT_UPLOAD_OPT   = '— Upload new file —'
_EXT_NO_FILES_OPT = '— no audio files found —'
_EXT_SEAM_OVERLAP = 0.5   # seconds of existing audio regenerated at the joint, so the splice is composed rather than butt-joined

def _ext_limit_html(model_name):
    cap = MODEL_CONFIGS[model_name]['max_duration']
    return (
        '<div style="font-size:12px;color:#4B5563;background:#F3F4F6;'
        'border-left:3px solid #9CA3AF;padding:6px 10px;border-radius:3px;margin:2px 0;">'
        f'ℹ️  <b>{model_name}</b> caps the <i>total</i> output at '
        f'<b>{cap:.0f} s</b> — source + extension must fit within it.'
        '</div>'
    )

# ── State ─────────────────────────────────────────────────────────────────────
_ext_audio = None   # Tensor (channels, samples) at SAMPLE_RATE
_ext_label = None   # display name of the loaded source

# ── Drive helpers ─────────────────────────────────────────────────────────────
def _ext_scan_drive():
    if not os.path.isdir(_EXT_DRIVE_FOLDER):
        return {}
    return {
        fname: os.path.join(_EXT_DRIVE_FOLDER, fname)
        for fname in sorted(os.listdir(_EXT_DRIVE_FOLDER))
        if os.path.splitext(fname)[1].lower() in _EXT_AUDIO_EXTS
    }

_ext_drive_files = _ext_scan_drive()
_ext_drive_ok    = bool(_ext_drive_files)

# ── Widgets ───────────────────────────────────────────────────────────────────
ext_model_dd = widgets.Dropdown(
    options=[(MODEL_CONFIGS[k]["label"], k) for k in MODEL_CONFIGS],
    value="small-music",
    description="Model:",
    style={"description_width": "52px"},
    layout=Layout(width="500px"),
)
ext_gpu_warn = widgets.HTML(
    value='<div class="sa3-warn">⚠️  <b>medium</b> requires a GPU. '
          'Go to <b>Runtime → Change runtime type → GPU</b>.</div>',
    layout=Layout(display="none", width="100%"),
)
ext_limit_note = widgets.HTML(
    value=_ext_limit_html("small-music"), layout=Layout(width="100%")
)

# Audio source
ext_source_dd = widgets.Dropdown(
    options=[_EXT_UPLOAD_OPT], description="Source:",
    style={"description_width": "60px"}, layout=Layout(width="440px"),
)
ext_upload_btn = widgets.Button(
    description="Upload Audio", icon="upload",
    layout=Layout(width="155px", height="36px"),
)
ext_upload_btn.style.button_color = "#EDE9FE"
ext_upload_status = widgets.HTML(
    value='<div class="sa3-info" style="margin:4px 0;">No file loaded yet.</div>'
)
ext_upload_row = widgets.HBox(
    [ext_upload_btn, ext_upload_status],
    layout=Layout(align_items="center", gap="10px", margin="4px 0 0"),
)

_ext_dstat_val = (
    f'{len(_ext_drive_files)} file(s) found' if _ext_drive_ok
    else ('Drive not mounted or folder not found'
          if not os.path.isdir(_EXT_DRIVE_FOLDER) else 'No audio files in folder')
)
ext_drive_dd = widgets.Dropdown(
    options=list(_ext_drive_files) if _ext_drive_ok else [_EXT_NO_FILES_OPT],
    description="", layout=Layout(width="360px"),
)
ext_drive_load_btn = widgets.Button(
    description=" Load", icon="folder-open",
    layout=Layout(width="100px", height="34px"), disabled=not _ext_drive_ok,
)
ext_drive_load_btn.style.button_color = "#EDE9FE"
ext_drive_refresh_btn = widgets.Button(
    description="", icon="refresh",
    layout=Layout(width="36px", height="34px"), tooltip="Re-scan folder",
)
ext_drive_status = widgets.HTML(
    value=f'<span style="font-size:12px;color:#9CA3AF;">{_ext_dstat_val}</span>'
)
ext_drive_row = widgets.HBox(
    [ext_drive_dd, ext_drive_load_btn, ext_drive_refresh_btn, ext_drive_status],
    layout=Layout(align_items="center", gap="6px"),
)
ext_audio_out = widgets.Output()

# Mode
ext_mode = widgets.ToggleButtons(
    options=[("Continue end", "end"), ("Extend intro", "intro"), ("Seamless loop", "loop")],
    value="end",
    style={"button_width": "150px"},
)
ext_mode_guide = widgets.HTML(value=(
    '<div style="font-size:12px;color:#4B5563;margin:4px 0 8px;line-height:1.6;">'
    '<b>Continue end</b> — new material after the clip, composed to follow from it. '
    '<b>Extend intro</b> — a lead-in composed to arrive at the clip. '
    '<b>Seamless loop</b> — the start/end joint is regenerated so the clip repeats without a bump. '
    f'In every mode the model also re-writes {_EXT_SEAM_OVERLAP:.1f}&thinsp;s of existing '
    'audio at the joint so the splice is played, not glued.'
    '</div>'
))

# Extension length (extend modes)
ext_len_slider = widgets.FloatSlider(
    value=10.0, min=1.0, max=60.0, step=0.5,
    description="", layout=Layout(width="490px"),
    readout_format=".1f", continuous_update=False,
)
ext_len_box = widgets.VBox([
    widgets.HTML(
        '<div style="font-size:12px;color:#6B7280;margin:8px 0 1px;">'
        '<b>Extension</b> '
        '<span style="color:#9CA3AF;">— seconds of new audio to add (capped by the model max)</span></div>'
        '<div style="display:flex;justify-content:space-between;width:490px;'
        'font-size:11px;color:#9CA3AF;margin:1px 0 2px;">'
        '<span>1 — a fill</span><span>60 — a new section</span></div>'
    ),
    ext_len_slider,
])

# Seam window (loop mode)
ext_seam_slider = widgets.FloatSlider(
    value=3.0, min=0.5, max=8.0, step=0.25,
    description="", layout=Layout(width="490px"),
    readout_format=".2f", continuous_update=False,
)
ext_seam_box = widgets.VBox([
    widgets.HTML(
        '<div style="font-size:12px;color:#6B7280;margin:8px 0 1px;">'
        '<b>Seam window</b> '
        '<span style="color:#9CA3AF;">— seconds regenerated around the loop joint</span></div>'
        '<div style="display:flex;justify-content:space-between;width:490px;'
        'font-size:11px;color:#9CA3AF;margin:1px 0 2px;">'
        '<span>0.5 — surgical</span><span>8 — recomposed transition</span></div>'
    ),
    ext_seam_slider,
])
ext_seam_box.layout.display = "none"

# Prompts
ext_prompt = widgets.Textarea(
    value="",
    placeholder="Optional — describe the new material (e.g. “builds into a euphoric chorus”). "
                "Leave blank to let the context decide.",
    layout=Layout(width="100%", height="56px"),
    style={"description_width": "0px"},
)
ext_neg_prompt = widgets.Textarea(
    value="",
    placeholder="Optional negative prompt — what to avoid (e.g. “distortion, noise”)",
    layout=Layout(width="100%", height="44px"),
    style={"description_width": "0px"},
)

# CFG
ext_cfg_slider = widgets.FloatSlider(
    value=1.0, min=0.0, max=15.0, step=0.5,
    description="", layout=Layout(width="490px"),
    readout_format=".1f", continuous_update=False,
)
ext_cfg_guide = widgets.HTML(value=(
    '<div style="font-size:12px;color:#4B5563;margin:2px 0 8px;line-height:1.6;">'
    'CFG (Classifier-Free Guidance) — how strongly the new region follows the prompt. '
    '<b>1.0</b> = blend seamlessly with the surrounding audio; <b>2–4</b> = clearly follow '
    'the prompt, with some risk of a stylistic jump at the joint.'
    '</div>'
))

# Steps
ext_steps_slider = widgets.IntSlider(
    value=8, min=4, max=50, step=1,
    description="", layout=Layout(width="490px"),
    continuous_update=False,
)
ext_steps_guide = widgets.HTML(value=(
    '<div style="font-size:12px;color:#4B5563;margin:2px 0 8px;line-height:1.6;">'
    'Denoising steps &mdash; the model is tuned for <b>8</b>. More steps refine texture '
    'with diminishing returns; cost grows roughly linearly (16 steps &asymp; 2&times; the time).'
    '</div>'
))

# Seed
ext_seed = widgets.BoundedIntText(
    value=-1, min=-1, max=2**31-1,
    description="Seed:", style={"description_width": "40px"},
    layout=Layout(width="160px"),
)

# Generate
ext_btn = widgets.Button(
    description="  Generate", icon="magic", button_style="",
    layout=Layout(width="150px", height="42px"),
)
ext_btn.style.button_color = "#7C3AED"
ext_btn.style.font_weight  = "700"
ext_out = widgets.Output(layout=Layout(min_height="60px", margin="6px 0 0"))


# ── Helpers ───────────────────────────────────────────────────────────────────
def _ext_load(audio_tensor, label):
    global _ext_audio, _ext_label
    _ext_audio = audio_tensor
    _ext_label = label
    dur = audio_tensor.shape[-1] / SAMPLE_RATE
    ext_upload_status.value = (
        f'<div class="sa3-info" style="margin:4px 0;">✅  '
        f'<strong>{label}</strong>  |  {dur:.1f} s '
        f' |  {audio_tensor.shape[0]} ch</div>'
    )
    _tmp = '/tmp/_sa3_ext_src.wav'
    torchaudio.save(_tmp, audio_tensor.cpu(), SAMPLE_RATE)
    with ext_audio_out:
        clear_output(wait=True)
        display(Audio(_tmp, autoplay=False))


# ── Callbacks ─────────────────────────────────────────────────────────────────
def _ext_on_model(change):
    cfg = MODEL_CONFIGS[change["new"]]
    ext_gpu_warn.layout.display = "" if cfg["needs_gpu"] and not torch.cuda.is_available() else "none"
    ext_limit_note.value = _ext_limit_html(change["new"])

def _ext_on_mode(change):
    is_loop = change["new"] == "loop"
    ext_len_box.layout.display  = "none" if is_loop else ""
    ext_seam_box.layout.display = "" if is_loop else "none"

def _ext_on_source(change):
    title = change["new"]
    if title == _EXT_UPLOAD_OPT:
        ext_upload_row.layout.display = ""
    else:
        ext_upload_row.layout.display = "none"
        audio, sr = torchaudio.load(_session_audio[title])
        if sr != SAMPLE_RATE:
            audio = torchaudio.functional.resample(audio, sr, SAMPLE_RATE)
        _ext_load(audio, title)

def _ext_on_upload(b):
    uploaded = colab_files.upload()
    for fname, data in uploaded.items():
        tmp = f'/tmp/{fname}'
        with open(tmp, 'wb') as f:
            f.write(data)
        audio, sr = torchaudio.load(tmp)
        if sr != SAMPLE_RATE:
            audio = torchaudio.functional.resample(audio, sr, SAMPLE_RATE)
        _ext_load(audio, fname)

def _ext_on_drive_load(b):
    fname = ext_drive_dd.value
    if fname not in _ext_drive_files:
        return
    ext_drive_status.value = '<span style="font-size:12px;color:#6B7280;">Loading…</span>'
    try:
        audio, sr = torchaudio.load(_ext_drive_files[fname])
        if sr != SAMPLE_RATE:
            audio = torchaudio.functional.resample(audio, sr, SAMPLE_RATE)
        _ext_load(audio, fname)
        ext_drive_status.value = f'<span style="font-size:12px;color:#059669;">✓ {fname}</span>'
    except Exception as exc:
        ext_drive_status.value = f'<div class="sa3-error">Error: {exc}</div>'

def _ext_on_drive_refresh(b):
    global _ext_drive_files, _ext_drive_ok
    _ext_drive_files = _ext_scan_drive()
    _ext_drive_ok = bool(_ext_drive_files)
    if _ext_drive_ok:
        ext_drive_dd.options = list(_ext_drive_files)
        ext_drive_load_btn.disabled = False
        ext_drive_status.value = f'<span style="font-size:12px;color:#6B7280;">{len(_ext_drive_files)} file(s) found</span>'
    else:
        ext_drive_dd.options = [_EXT_NO_FILES_OPT]
        ext_drive_load_btn.disabled = True
        _msg = ("Drive not mounted or folder not found"
                if not os.path.isdir(_EXT_DRIVE_FOLDER) else "No audio files in folder")
        ext_drive_status.value = f'<span style="font-size:12px;color:#9CA3AF;">{_msg}</span>'

def _ext_on_generate(b):
    if _ext_audio is None:
        with ext_out:
            clear_output(wait=True)
            display(HTML('<div class="sa3-error">Select or upload an audio file first.</div>'))
        return

    model_name = ext_model_dd.value
    mode       = ext_mode.value
    prompt     = ext_prompt.value.strip() or None
    neg_prompt = ext_neg_prompt.value.strip() or None
    cfg        = ext_cfg_slider.value
    steps      = ext_steps_slider.value
    seed       = ext_seed.value
    cap        = MODEL_CONFIGS[model_name]['max_duration']
    src        = _ext_audio
    src_dur    = src.shape[-1] / SAMPLE_RATE

    # ── Mode-specific validation and mask/audio construction ──────────────────
    if mode in ("end", "intro"):
        if src_dur >= cap - 1.0:
            with ext_out:
                clear_output(wait=True)
                display(HTML(
                    f'<div class="sa3-error">The source is already {src_dur:.0f}&thinsp;s and '
                    f'<b>{model_name}</b> caps the total at {cap:.0f}&thinsp;s — there is no room '
                    f'to extend. Trim the source or switch model.</div>'
                ))
            return
        target  = min(src_dur + ext_len_slider.value, cap)
        pad_sec = target - src_dur
        pad = torch.zeros(src.shape[0], int(round(pad_sec * SAMPLE_RATE)), dtype=src.dtype)
        if mode == "end":
            padded     = torch.cat([src, pad], dim=-1)
            mask_start = max(0.0, src_dur - _EXT_SEAM_OVERLAP)
            mask_end   = target
        else:
            padded     = torch.cat([pad, src], dim=-1)
            mask_start = 0.0
            mask_end   = min(pad_sec + _EXT_SEAM_OVERLAP, target)
        gen_duration = target
        shift = 0
    else:  # loop
        if src_dur > cap:
            with ext_out:
                clear_output(wait=True)
                display(HTML(
                    f'<div class="sa3-error">The source is {src_dur:.0f}&thinsp;s but '
                    f'<b>{model_name}</b> processes at most {cap:.0f}&thinsp;s — trim the clip '
                    f'first so the loop stays sample-exact.</div>'
                ))
            return
        if src_dur < 4.0:
            with ext_out:
                clear_output(wait=True)
                display(HTML('<div class="sa3-error">Loop mode needs at least a 4&thinsp;s clip.</div>'))
            return
        # Rotate so the start/end joint sits mid-clip, where we can inpaint it.
        shift  = src.shape[-1] // 2
        padded = torch.roll(src, shifts=shift, dims=-1)
        half_w = min(ext_seam_slider.value, src_dur - 1.0) / 2.0
        center = src_dur / 2.0
        mask_start = max(0.25, center - half_w)
        mask_end   = min(src_dur - 0.25, center + half_w)
        gen_duration = src_dur

    ext_btn.disabled    = True
    ext_btn.description = "  Generating…"

    if model_name in _model_cache:
        _st = {"end": "Composing the continuation…", "intro": "Composing the intro…",
               "loop": "Regenerating the loop joint…"}[mode]
        _ss = f"⚡  <strong>{model_name}</strong> already loaded."
    elif _is_model_on_disk(model_name):
        _st = f"Loading <strong>{model_name}</strong> from cache…"
        _ss = "💾  Weights found on Drive — no download needed."
    else:
        _st = f"Downloading <strong>{model_name}</strong>…"
        _ss = "⬇  First run — may take 1–3 min."

    with ext_out:
        clear_output(wait=True)
        display(HTML(SPINNER_HTML.format(status=_st, substatus=_ss)))
    try:
        import sys as _sys
        _rec = _sys.getrecursionlimit()
        _sys.setrecursionlimit(max(_rec, 4000))
        try:
            sa3 = get_model(model_name)
        finally:
            _sys.setrecursionlimit(_rec)

        # If the model was cached from a CPU session but GPU is now
        # available, transparently reload it on the GPU.
        import gc as _gc2
        _want = "cuda" if torch.cuda.is_available() else "cpu"
        _have = str(next(sa3.model.parameters()).device)
        if not _have.startswith(_want):
            del _model_cache[model_name]
            _gc2.collect()
            if torch.cuda.is_available():
                torch.cuda.empty_cache()
            with ext_out:
                clear_output(wait=True)
                display(HTML(SPINNER_HTML.format(
                    status=f"Reloading <strong>{model_name}</strong> on GPU…",
                    substatus="⚡ Model was cached on CPU — reloading on CUDA.",
                )))
            _sys.setrecursionlimit(max(_sys.getrecursionlimit(), 4000))
            sa3 = get_model(model_name)
            _have = str(next(sa3.model.parameters()).device)
        _dev_label = "GPU ✓" if _have.startswith("cuda") else "CPU"

        t0 = time.perf_counter()
        with torch.inference_mode():
            generated = sa3.generate(
                prompt=prompt,
                negative_prompt=neg_prompt,
                inpaint_audio=(SAMPLE_RATE, padded),
                inpaint_mask_start_seconds=mask_start,
                inpaint_mask_end_seconds=mask_end,
                duration=gen_duration,
                steps=steps,
                cfg_scale=cfg,
                seed=seed,
                chunked_decode=True,
            )
        elapsed = time.perf_counter() - t0

        result = generated[0].cpu()
        if mode == "loop":
            # Match the rotated input's length exactly, then rotate back so the
            # regenerated joint lands at the clip's start/end boundary.
            n = padded.shape[-1]
            if result.shape[-1] > n:
                result = result[..., :n]
            elif result.shape[-1] < n:
                result = torch.nn.functional.pad(result, (0, n - result.shape[-1]))
            result = torch.roll(result, shifts=-shift, dims=-1)

        out_path = '/tmp/_sa3_ext_result.wav'
        torchaudio.save(out_path, result, SAMPLE_RATE)
        _title_prefix = {"end": "Continued", "intro": "Intro added", "loop": "Loop"}[mode]
        _reg_title = _register_session_clip(f"{_title_prefix}: {(_ext_label or 'clip')[:40]}", out_path)

        _x2_path = None
        if mode == "loop":
            _x2_path = '/tmp/_sa3_ext_loop_x2.wav'
            torchaudio.save(_x2_path, torch.cat([result, result], dim=-1), SAMPLE_RATE)

        _dl_btn = widgets.Button(
            description="  Download", icon="download",
            layout=Layout(width="150px", height="34px"),
        )
        _dl_btn.style.button_color = "#EDE9FE"
        def _do_dl(b, _p=out_path):
            from google.colab import files as _cf
            _cf.download(_p)
        _dl_btn.on_click(_do_dl)

        prompt_short = ((prompt or "")[:90] + "…") if prompt and len(prompt) > 90 else (prompt or "")
        _mode_label  = {"end": "continue end", "intro": "extend intro", "loop": "seamless loop"}[mode]
        _region_stat = (f"seam {mask_end - mask_start:.2f}s" if mode == "loop"
                        else f"+{gen_duration - src_dur:.1f}s → {gen_duration:.1f}s total")

        with ext_out:
            clear_output(wait=True)
            display(HTML(f"""
            <div class="sa3-result">
              <div style="display:flex;align-items:center;gap:8px;margin-bottom:8px;flex-wrap:wrap;">
                <span class="sa3-badge">{model_name}</span>
                <span class="sa3-stat">{_dev_label}</span>
                <span class="sa3-stat">{_mode_label}</span>
                <span class="sa3-stat">{_region_stat}</span>
                <span class="sa3-stat">CFG {cfg:.1f}</span>
                <span class="sa3-stat">generated in {elapsed:.1f} s</span>
              </div>
              {'<div style="font-size:13px;color:#374151;font-style:italic;line-height:1.5;">&ldquo;' + prompt_short + '&rdquo;</div>' if prompt_short else ''}
            </div>
            """))
            display(Audio(out_path, autoplay=True))
            if _x2_path is not None:
                display(HTML(
                    '<div style="font-size:12px;color:#6B7280;margin:8px 0 2px;">'
                    '<b>Seam check</b> — the loop played twice back to back; '
                    'the joint passes at the midpoint:</div>'
                ))
                display(Audio(_x2_path, autoplay=False))
            display(_dl_btn)
            display(HTML(f'<div style="font-size:12px;color:#9CA3AF;margin:6px 0 0;">Saved to session clips as &ldquo;<b>{_reg_title}</b>&rdquo; &mdash; available as a source in every tool.</div>'))
    except Exception as exc:
        with ext_out:
            clear_output(wait=True)
            display(HTML(f'<div class="sa3-error"><strong>Error:</strong> {exc}</div>'))
    finally:
        ext_btn.disabled    = False
        ext_btn.description = "  Generate"


# ── Connect ───────────────────────────────────────────────────────────────────
ext_model_dd.observe(_ext_on_model, names="value")
ext_mode.observe(_ext_on_mode, names="value")
ext_source_dd.observe(_ext_on_source, names="value")
ext_upload_btn.on_click(_ext_on_upload)
ext_drive_load_btn.on_click(_ext_on_drive_load)
ext_drive_refresh_btn.on_click(_ext_on_drive_refresh)
ext_btn.on_click(_ext_on_generate)

def _ext_refresh_sources():
    prev = ext_source_dd.value
    opts = [_EXT_UPLOAD_OPT] + list(_session_audio.keys())
    ext_source_dd.options = opts
    if prev in opts:
        ext_source_dd.value = prev
_add_session_listener('extend', _ext_refresh_sources)
if _session_audio:
    ext_source_dd.options = [_EXT_UPLOAD_OPT] + list(_session_audio.keys())

# ── Layout ────────────────────────────────────────────────────────────────────
display(widgets.VBox([
    widgets.HTML('<div style="font-size:12px;color:#6B7280;margin:0 0 3px;">Model:</div>'),
    ext_model_dd,
    ext_gpu_warn,
    ext_limit_note,
    widgets.HTML('<div style="font-size:12px;color:#6B7280;margin:8px 0 3px;">Audio source:</div>'),
    ext_source_dd,
    ext_upload_row,
    widgets.HTML(
        f'<div style="font-size:12px;color:#6B7280;margin:6px 0 2px;">'
        f'Drive folder — <code>My Drive/{os.path.basename(_EXT_DRIVE_FOLDER)}</code>:</div>'
    ),
    ext_drive_row,
    widgets.HTML('<div style="font-size:12px;color:#6B7280;margin:6px 0 2px;">Source audio:</div>'),
    ext_audio_out,
    widgets.HTML('<div style="font-size:12px;color:#6B7280;margin:10px 0 3px;">Mode:</div>'),
    ext_mode,
    ext_mode_guide,
    ext_len_box,
    ext_seam_box,
    widgets.HTML(
        '<div style="font-size:12px;color:#6B7280;margin:10px 0 2px;">'
        'Prompt <span style="color:#9CA3AF;">— optional; describes only the regenerated region</span></div>'
    ),
    ext_prompt,
    widgets.HTML(
        '<div style="font-size:12px;color:#6B7280;margin:6px 0 2px;">'
        'Negative prompt <span style="color:#9CA3AF;">— optional</span></div>'
    ),
    ext_neg_prompt,
    widgets.HTML(
        '<div style="font-size:12px;color:#6B7280;margin:10px 0 1px;">'
        '<b>Prompt strength</b> '
        '<span style="color:#9CA3AF;">— CFG scale (ignored without a prompt)</span></div>'
        '<div style="display:flex;justify-content:space-between;width:490px;'
        'font-size:11px;color:#9CA3AF;margin:1px 0 2px;">'
        '<span>0 — ignore prompt</span><span>15 — follow strictly</span></div>'
    ),
    ext_cfg_slider,
    ext_cfg_guide,
    widgets.HTML(
        '<div style="font-size:12px;color:#6B7280;margin:8px 0 1px;">'
        '<b>Steps</b> '
        '<span style="color:#9CA3AF;">&mdash; denoising steps (default 8)</span></div>'
        '<div style="display:flex;justify-content:space-between;width:490px;'
        'font-size:11px;color:#9CA3AF;margin:1px 0 2px;">'
        '<span>4 &mdash; fastest</span><span>50 &mdash; max refinement</span></div>'
    ),
    ext_steps_slider,
    ext_steps_guide,
    ext_seed,
    ext_btn,
    ext_out,
], layout=Layout(width="100%", max_width="700px")))
```

```python
#@title 11 — Clip Inspector

display(HTML("""
<div class="sa3-section">
  <h2>11 &mdash; Clip Inspector</h2>
  <p>
    Look at what you just made. Pick any session clip (or upload one) and it is analysed
    instantly &mdash; waveform, spectrogram, and level statistics &mdash; so you can compare
    takes and spot clipping, dead air, or missing high end without leaving the notebook.
    No model involved: this cell runs in a second on CPU.
  </p>
</div>
"""))

# ── Imports (other names come from cells 4–5) ──────────────────────────────────
from google.colab import files as colab_files

_INSP_UPLOAD_OPT = '— Upload new file —'
_INSP_PICK_OPT   = '— pick a clip —'

# ── Widgets ───────────────────────────────────────────────────────────────────
insp_source_dd = widgets.Dropdown(
    options=[_INSP_PICK_OPT, _INSP_UPLOAD_OPT], description="Clip:",
    style={"description_width": "60px"}, layout=Layout(width="440px"),
)
insp_upload_btn = widgets.Button(
    description="Upload Audio", icon="upload",
    layout=Layout(width="155px", height="36px"),
)
insp_upload_btn.style.button_color = "#EDE9FE"
insp_upload_row = widgets.HBox(
    [insp_upload_btn],
    layout=Layout(align_items="center", gap="10px", margin="4px 0 0"),
)
insp_upload_row.layout.display = "none"
insp_out = widgets.Output(layout=Layout(min_height="40px", margin="6px 0 0"))


# ── Analysis ──────────────────────────────────────────────────────────────────
def _insp_db(x, floor=-120.0):
    import math
    return floor if x <= 0 else max(floor, 20.0 * math.log10(x))

def _insp_analyze(audio, label):
    """audio: Tensor (channels, samples) at SAMPLE_RATE."""
    import numpy as np
    import matplotlib.pyplot as plt

    n_ch  = audio.shape[0]
    n     = audio.shape[-1]
    dur   = n / SAMPLE_RATE
    mono  = audio.mean(dim=0).numpy()
    peak  = float(audio.abs().max())
    rms   = float((audio ** 2).mean() ** 0.5)
    peak_db, rms_db = _insp_db(peak), _insp_db(rms)
    crest = peak_db - rms_db
    # Fraction of samples within 0.5 dB of full scale — a quick clipping smell test
    clip_frac = float((audio.abs() > 0.944).float().mean())

    with insp_out:
        clear_output(wait=True)
        display(HTML(f"""
        <div class="sa3-result">
          <div style="display:flex;align-items:center;gap:8px;margin-bottom:4px;flex-wrap:wrap;">
            <span class="sa3-badge">{label[:60]}</span>
            <span class="sa3-stat">{dur:.1f}&thinsp;s</span>
            <span class="sa3-stat">{n_ch}&thinsp;ch · {SAMPLE_RATE // 1000}&thinsp;kHz</span>
            <span class="sa3-stat">peak {peak_db:.1f}&thinsp;dBFS</span>
            <span class="sa3-stat">RMS {rms_db:.1f}&thinsp;dBFS</span>
            <span class="sa3-stat">crest {crest:.1f}&thinsp;dB</span>
          </div>
          {'<div class="sa3-warn" style="margin:6px 0 0;">⚠️&nbsp; ' + f'{clip_frac * 100:.2f}% of samples sit at full scale — likely clipping. Consider normalizing down in a DAW.' + '</div>' if clip_frac > 0.0005 else ''}
        </div>
        """))
        _tmp = '/tmp/_sa3_insp_src.wav'
        torchaudio.save(_tmp, audio.cpu(), SAMPLE_RATE)
        display(Audio(_tmp, autoplay=False))

        fig, (ax_w, ax_s) = plt.subplots(
            2, 1, figsize=(9.5, 5.2), height_ratios=[1, 2],
            constrained_layout=True,
        )
        # Waveform (decimated for speed; peak-preserving envelope)
        step = max(1, n // 4000)
        seg  = mono[: (n // step) * step].reshape(-1, step)
        t    = np.arange(seg.shape[0]) * step / SAMPLE_RATE
        ax_w.fill_between(t, seg.min(axis=1), seg.max(axis=1),
                          color='#7C3AED', alpha=0.85, linewidth=0)
        ax_w.set_xlim(0, dur)
        ax_w.set_ylim(-1.05, 1.05)
        ax_w.set_ylabel('amplitude', fontsize=8)
        ax_w.tick_params(labelsize=7)
        ax_w.set_title('Waveform (mono mix)', fontsize=9, color='#5B21B6', loc='left')
        for spine in ax_w.spines.values():
            spine.set_color('#DDD6FE')
        # Spectrogram
        ax_s.specgram(mono, NFFT=2048, Fs=SAMPLE_RATE, noverlap=1024,
                      cmap='magma', vmin=-120, vmax=0)
        ax_s.set_xlim(0, dur)
        ax_s.set_ylabel('Hz', fontsize=8)
        ax_s.set_xlabel('seconds', fontsize=8)
        ax_s.tick_params(labelsize=7)
        ax_s.set_title('Spectrogram (dB)', fontsize=9, color='#5B21B6', loc='left')
        for spine in ax_s.spines.values():
            spine.set_color('#DDD6FE')
        plt.show()


# ── Callbacks ─────────────────────────────────────────────────────────────────
def _insp_on_source(change):
    title = change["new"]
    if title == _INSP_PICK_OPT:
        return
    if title == _INSP_UPLOAD_OPT:
        insp_upload_row.layout.display = ""
        return
    insp_upload_row.layout.display = "none"
    try:
        audio, sr = torchaudio.load(_session_audio[title])
        if sr != SAMPLE_RATE:
            audio = torchaudio.functional.resample(audio, sr, SAMPLE_RATE)
        _insp_analyze(audio, title)
    except Exception as exc:
        with insp_out:
            clear_output(wait=True)
            display(HTML(f'<div class="sa3-error"><strong>Error:</strong> {exc}</div>'))

def _insp_on_upload(b):
    uploaded = colab_files.upload()
    for fname, data in uploaded.items():
        tmp = f'/tmp/{fname}'
        with open(tmp, 'wb') as f:
            f.write(data)
        audio, sr = torchaudio.load(tmp)
        if sr != SAMPLE_RATE:
            audio = torchaudio.functional.resample(audio, sr, SAMPLE_RATE)
        _insp_analyze(audio, fname)

insp_source_dd.observe(_insp_on_source, names="value")
insp_upload_btn.on_click(_insp_on_upload)

def _insp_refresh_sources():
    prev = insp_source_dd.value
    opts = [_INSP_PICK_OPT, _INSP_UPLOAD_OPT] + list(_session_audio.keys())
    insp_source_dd.options = opts
    if prev in opts:
        insp_source_dd.value = prev
_add_session_listener('inspect', _insp_refresh_sources)
if _session_audio:
    insp_source_dd.options = [_INSP_PICK_OPT, _INSP_UPLOAD_OPT] + list(_session_audio.keys())

# ── Layout ────────────────────────────────────────────────────────────────────
display(widgets.VBox([
    widgets.HTML(
        '<div style="font-size:12px;color:#6B7280;margin:0 0 3px;">'
        'Pick a clip &mdash; it is analysed the moment you select it:</div>'
    ),
    insp_source_dd,
    insp_upload_row,
    insp_out,
], layout=Layout(width="100%", max_width="700px")))
```
